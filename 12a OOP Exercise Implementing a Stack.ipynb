**Introduction**

Recall from the lecture, that a stack is a Last In First Out (LIFO) store that finds many uses in computing, one of the most obvious is the ability to remember and move backwards and forwards in an internet environment or in the VLE for this course. The address of the last place viewed is pushed onto the stack when we move forward and then retrieved from the stack as we move backwards. Two steps back removes the last two addresses. 


&emsp;&emsp;&emsp;<img src="images\The_Stack.png" width="450"/>



When we divert from the path of a linear prorgram, to execute a function or subroutine, how do we know where to return to in the code? The answer involves the stack, it has the address of the next instruction pushed on to a stack at the moment it diverts to the function. When the code returns from the function, it pulls that address from the stack and continues on its way. 

The stack is a fundamental **data structure**, and a system like windows may be running thousands of stacks at any one time, a typical size is of the order of 1 MByte.

We have seen how a stack can be implemented with procedural code using a list and append instructions but the difficulties arose when multiple stacks were needed. Renaming the stacks and their functions became tedious and in addition:

>It could be compromised  
>It could be overwritten with careless code   
>Two stacks would require two lists and two new push and pop instructions with different names or  
> More intricate code to deal with a fixed number of stacks  

What if the number of stacks was undefined at the time of writing?

As a reminder, a procedural stack implementation may look like the following, try and predict the outcome of the code and then run it:


```python
# Example - review creating a procedural stack

stack = []

def push(val):
    stack.append(val)  # Why not use insert?

def pop():
    val = stack[-1]
    del stack[-1]
    return val

# main
# Push in order 1, 2, 3
push(1) #first in
push(2)
push(3) #last in

# Pop in order 3, 2, 1
print(pop()) #first out
print(pop())
print(pop()) #last out
```

**Privacy**

To prevent unauthorised or accidnetal access to our newly created stack, we need to make it private or at least the stack_list[ ] private.  

You will find mixed feelings on how best to do this. Python phlosphy would argue there should be no private variables:

&emsp;&emsp;&emsp;&emsp;"Always decide whether a class’s methods and instance variables (collectively: “attributes”) should be public or non-public. If in doubt, choose non-public; it’s easier to make it public later than to make a public attribute non-public."

Public attributes are those that you expect unrelated clients of your class to use.  
Non-public attributes are those that are not intended to be used by third parties; you make no guarantees that non-public attributes won’t change or even be removed.

&emsp;&emsp;*We don’t use the term “private” here, since no attribute is really private in Python (without a generally unnecessary amount of work)." - PEP8*


Well if varaibles can't be made truely private how are they made non-public?

It is encouraged that a single underscore, e.g. _var be added to a variable inorder to indicate to users that a variable should not be accessed. 

> _my_variable   
> _colour  
> _position   

etc.

This doesn't stop it being accessed but it signals the intent.

Alternatively, a double underscore makes it much more difficult to 'accidently' access variables with same names e.g. __var. When this is detected, python executes a name mangleing routine that adds the class name to the front of the variable, thuse **__var** in class Stack would become: **_Stack__var**. This would be much more difficult to get 'accidently' but doesn't stop direct access for the determined.

This is how we encapsulate the data within our classes and prevent external access except through **getters** and **setters**.

Remember:

Single Underscore (_)   
>Convention for “protected” attributes: Signals that the attribute is intended to be private, but can still be accessed if needed.  
Does not enforce privacy: It's more of a gentle warning to developers not to modify the attribute directly.  

Double Underscore (__) (Name Mangling)   
>Strongly discourages direct access: Python mangles the name to _ClassName__attribute, making external access harder.  
Not truly private: Can still be accessed with name mangling (_Stack__stack_list).
