# List comprehension

Consider, a list that contains 5 numbers and you want to add 1 (+1) to each of the numbers in the list; this could be achieved in the following way:

```python
numbers = [1, 2, 3, 4, 5]

new_list = []
for n in numbers:
    add_1 = n + 1
    new_list.append(add_1)

print("Ammended list = ", new_list)
```

You should see that this code achieves exactly what we set out to do, that is add 1 to each of the numbers in the list. This is a perfectly fair bit of code but python provides alternatives; test the code below:

```python
numbers = [1, 2, 3, 4, 5]

new_list = [n+1  for n in numbers]

print("Amended list = ", new_list)
```

This achieves exactly the same in a more compact form; the new_list is generated from a:

>&emsp;**List Comprehension**

List comprehensions in Python are an elegant and concise way to create and manipulate lists. They enable you to write compact code that generates a new list based on an existing sequence or iterable, often in just one line.

Key Features:

>&emsp;**Readable:** List comprehensions are more readable than traditional loops.  
&emsp;**Efficient:** They tend to execute faster than manually appending elements in a loop.  
&emsp;**Flexible:** You can include conditional statements to filter elements  

The syntax is as follows:

>&emsp;new_list = **[new_item  for n in old_list]**  

or which can involve conditions as follows:

>&emsp;new_list = **[expression for item in iterable if condition]**  

where:  

>&emsp;**Expression:** Defines the value to be added to the new list  
&emsp;**Item:** The variable that represents each element in the iterable  
&emsp;**Condition (Optional):** Filters which elements are included  

You will see how this translates to practical applications over the next examples:  

```python
# creating a new list of squares upto 10
squares = [x**2 for x in range(1, 11)]
print("Squares from 1 to 10", squares)

# Creating a list of even numbers
even_numbers = [x for x in range(10) if x % 2 == 0]  # condition that x is even x%2==0
print("Even numbers", even_numbers)

# Filtering a list
words = ["cat", "elephant", "dog", "rabbit"]
long_words = [word for word in words if len(word) > 3]
print("Long words", long_words)
```

These work for all "sequences"

```python
# Nested example
pairs = [(x, y) for x in range(3) for y in range(3) if x != y]
print(pairs)
```

**all()**  

The **all()** function is a handy built-in Python utility. It checks whether all elements of an iterable (like a list, tuple, etc.) evaluate to **True**. If every element satisfies the condition,  **all()** returns True; otherwise, it returns False.  

&emsp;**Input:** An iterable (like a list of conditions or booleans).  
&emsp;**Output:** True if all items are True; False if at least one item is False.  

```python
conditions = [True, True, False]
print(all(conditions))  # Output: False

conditions = [True, True, True]
print(all(conditions))  # Output: True
```

```python
# Calculating primes
primes = [x for x in range(2, 51) if all(x % i != 0 for i in range(2, x-1))]
print(primes)

# where:
# x for x in range(2, 51) -> takes the value x in the range 2 to 50 but only if...
# if all(x % i != 0  is true for every thing in the range... i.e. x divided by every other number upto 50 does not equal 0 . This is a test for a prime number
# for i in range(2, x-1))

# Example Walkthrough: x = 7
# range(2, 6) -> [2, 3, 4, 5]
# Test: Is 7 % i != 0 for every ,all, i?
# Passes: So 7 is kept
```
