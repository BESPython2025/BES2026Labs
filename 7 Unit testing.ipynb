```python
# An alternative test loop

def assign_grade(score):
    """Returns the grade category for a given score."""
    if score < 0 or score > 100:
        return "Invalid score! Must be between 0 and 100."
    elif score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

# Example cases to test boundaries and partitions
test_scores = [-1, 0, 59, 60, 69, 70, 79, 80, 89, 90, 99, 100, 101]

for score in test_scores:
    print(f"Score {score} → Grade: {assign_grade(score)}")
```

**Why this may not be a good solution**

>This loop clearly does check expected outputs, but doesn't **automatically validate correctness** —it still relies on a person **reading** the output to confirm the logic works. Where the tests are far more complicated, this task could be difficult and prone to error.  
**unittest** automates the validation, ensuring that tests fail visibly if the function behaves unexpectedly.  
If the loop test above fails, you still have to find and register the failure in your mind.  
As the project grows, manually inspecting printed results becomes unmanageable.  
Using unittest ensures consistent, repeatable tests across many functions without human inspection.  
Consider how difficult it would be to interpret the loop test data for someone who did not write the original function.

**White Box Testing**  

White box testing covers several key types of tests, each designed to examine the internal workings of a program rather than just its inputs and outputs. Where possible they would test:  

1.&emsp;&emsp;**Statement coverage**:  
&emsp;&emsp;&emsp;&emsp;Ensures that every line of code executes at least once during testing.  
&emsp;&emsp;&emsp;&emsp;Helps identify dead code (lines that never run).  
&emsp;&emsp;&emsp;&emsp;Example: If a  statement has multiple branches, statement coverage checks whether each branch gets tested.   

2.&emsp;&emsp;**Branch coverage**  
&emsp;&emsp;&emsp;&emsp;Confirms that both true and false conditions of every decision (if, elif , else) are tested.   
&emsp;&emsp;&emsp;&emsp;Prevents logical errors where certain branches might never execute.  

```python
# Branch coverage

if x > 10:  
    print("Greater")  # Needs to be tested for both >10 and ≤10 cases  
else:  
    print("Smaller or Equal")

# Branch coverage requires tests where x = 11 (true path) AND x = 10 (false path).
```

3.&emsp;&emsp;**Condition coverage**    
&emsp;&emsp;&emsp;&emsp;Tests individual boolean expressions inside conditional statements.  
&emsp;&emsp;&emsp;&emsp;Ensures that every logical condition within a decision point is exercised.  

```python
# Conditional coverage

if (x > 10 and y < 5):  
    print("Valid")  

# then all permutations of test cases should be tested:   
# x > 10 AND y < 5  
# x <= 10 AND y < 5  
# x > 10 AND y >= 5  
# x <= 10 AND y >= 5  
```

4.&emsp;&emsp;**Path coverage**  

&emsp;&emsp;&emsp;&emsp;Covers all possible execution flows from start to finish.  
&emsp;&emsp;&emsp;&emsp;Important for functions with loops and multiple conditions.


```python
# Path coverage

def calculate(x):
    if x > 0:
        x *= 2
    if x < 10:
        x += 3
    return x


# Path coverage ensures every unique execution sequence runs, such as:
# (x = 5):  Executes both conditions
# (x = 15): Executes only the first condition
# (x = -5): Executes only the second condition
```

5.&emsp;&emsp;**Loop testing** 

&emsp;&emsp;&emsp;&emsp;Ensures loops function properly under different conditions (zero iterations, one iteration, multiple iterations).

```python
while x < 10:
    x += 2

# Loop tests must cover:
# Zero iterations (x = 10)
# One iteration (x = 9)
# Multiple iterations (x = 0)
```

6.&emsp;&emsp;**Fault injection testing** 

&emsp;&emsp;&emsp;&emsp;Introduces intentional errors to see how the system handles failures.  
&emsp;&emsp;&emsp;&emsp;Example: What happens if a function receives a string instead of a expected number int or float?  

```python
def divide(x, y):
    if y == 0:
        return "Error: Division by zero"
    return x / y

# Fault injection tests cases where:
# (y = 0) Division by zero
# (y = none) Unexpected type
```

**Summary for white box testing**

&emsp;&emsp;White box testing ensures that logic flows correctly under all conditions.  
&emsp;&emsp;It verifies decision-making (condition coverage), fault handling, and execution paths.  
&emsp;&emsp;Testing internal structure ensures robustness beyond simple input/output validation.  

Testing a complex class can be as demanding as writing it, but structured units make it manageable. Breaking it into small, focused tests ensures thorough coverage without overwhelming effort.


Recall from the lecture; this difficult and often specialised job is not always done by the originating coder.

**Originating Coder Writes Tests (Common in TDD):**

Test-Driven Development (TDD) – The same developer writes tests first before coding the actual implementation.  
Ensures tight feedback loops and immediate validation of functionality.  
Common in agile workflows, especially for individual developers or small teams.  

**Dedicated Testing Engineers or QA Teams Write Tests**

Larger organizations often have specialized test engineers or QA teams focused on writing automated and manual tests.  
They validate correctness independently from the developer, ensuring objectivity in test coverage.   
Common in enterprise-level projects where quality assurance is separated from development.   

**Hybrid Approach: Developers and Test Engineers Share Testing Responsibilities**

Developers write unit tests for functions and modules.  
QA teams handle **integration tests, end-to-end tests
