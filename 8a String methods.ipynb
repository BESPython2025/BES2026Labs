**capitalize()**

>Cleaning up text where 'capitalization' is inconsistent.

**find()**

```python
# The second parameter is the position at which the search starts.
my_string = "congratulate the typesetter"
index = my_string.find("t", 8)
print(index)

# Some random text!
the_text = """A variation of the ordinary lorem ipsum
text has been used in typesetting since the 1960s 
or earlier, when it was popularised by advertisements 
for Letraset transfer sheets. It was introduced to 
the Information Age in the mid-1980s by the Aldus Corporation, 
which employed it in graphics and word-processing templates
for its desktop publishing program PageMaker (from Wikipedia)"""

fnd = the_text.find('the')
# Find all the occurrences of the string "the"
while fnd != -1:
    print(fnd)
    fnd = the_text.find('the', fnd + 1)
```

**strip()**

**Removes** all leading (left) and lagging (right) **white spaces** from a string

```python
# Demonstrating the strip() method:
print("[" + "   aleph   ".strip() + "]")
```

**lstrip()**

left-strip - creates a new string by removing all the leading white spaces and other specified characters - it ignores Lagging spaces - it treats the characters in the brackets as separate and not as a substring. Thus '.org' using rstrip would look separately for  '., o, r, g '

```python
# lstrip() method - Leading strip()

print("[" + "      tau       ".lstrip() + "]")
print("www.cisco.com".lstrip("wc."))
print("pythoninstitute.org".lstrip(".org"))
print("org.orgpythoninstitute.org".lstrip("org."))

print()

url = "https://example.com"
clean_url = url.lstrip("https://")
# Output: "example.com"
print(clean_url) 
```

**rstrip()**  

right-strip - strips white space and other character patterns from the right side of the string. like lstrip, it treats the characters in the brackets as separate and not as a substring. Thus '.com' using rstrip would look separately for  '., c, o, m '

```python
# rstrip() - right strip

print("[" + "      tau       ".rstrip() + "]")
print("www.cisco.com".rstrip(".com"))
print("pythoninstitute.org".rstrip("e.org"))
```

As with many of the string functions, it can be used for:  

>Cleaning user input: Removes unnecessary spaces at the beginning of strings   
>Formatting text data: Useful when processing files or logs with inconsistent spacing   
>Handling specific prefixes: Can remove unwanted characters from the start of a string  


**replace()**
A versatile method which:

>Replaces the first string parameter with the second, as many times as is found   
>The three parameter version restricts the number of times the substitution can take place  

```python
# simple replacement of strings
text = "Hello world, world is amazing!"
new_text = text.replace("world", "Earth")

# Output: "Hello Earth, Earth is amazing!"
print(new_text) 

print()


# only two of the possible three strings are replaced
text = "apple apple apple"
new_text = text.replace("apple", "orange", 2)

# Output: "orange orange apple"
print(new_text)  

print()

# Using replace in a loop and with a dictionary
text = "Hello world!"
replacements = {"Hello": "Hi", "world": "Earth"}
for old, new in replacements.items():
    text = text.replace(old, new)
# Output: "Hi Earth!"
print(text) 

print()

# It is case sensitive!
text = "Hello hello HELLO"
new_text = text.replace("hello", "Hi")

# Output: "Hello Hi HELLO"
print(new_text)  
```

**rfind()**

Return the highest index in the string where substring sub is found   
>The one-, two-, and three-parameter rfind() instructions do nearly the same as find() but start their searches from the end of the string, hence the prefix r for right.   
>syntax is:&emsp;&emsp;  string.rfind(substring, start, end)

```python
# Demonstrating the rfind() method:
text = "apple banana cherry apple"
position = text.rfind("apple")  # Search only in first 15 characters

print(position)  # Output: 20 

# Searching in a range
text = "apple banana cherry apple"
position = text.rfind("apple", 0, 15)  # Search only in first 15 characters

print(position)  # Output: 0 (only finds first occurrence)

print("\ne-mail example")
email = 'userxyz.com@domain.com'
last_dot_pos = email.rfind('.', 1)
print('last_dot_pos = ', last_dot_pos)

tld_string = email[last_dot_pos:] 	#recall slicing, slice from index 18
print('tld_string = ',tld_string) 

if tld_string == ".com":
    print("Email matched")
else:
    print("Email not matched, tld:", tld_string)
```

**split()**  

>Splits the string, as delimited by white spaces, and builds a *list* of all the detected substrings  
>If the string is empty then the resulting list is empty too
>splits can be orced using another string s a separator (example 2 below)

```python
# Demonstrating the split() method:
print("phi       chi       psi    ".split())

print("phi       chi\npsi".split())

# \n is also considered white space

print("Q. 1/16.5".split("/")[1].strip())

```

What can be said about the escape characters '\n'?

In the second example:

>"Q. 1/16.5" is split into a list, using "/" as the separator   
>the index `[1]` takes the second of these two parts which becomes: 16.5  
>the strip cleans of any lead or lagging spaces - of which there are none 
>leaves the answer as 16.5

Introduction
The International Bank Account Number (IBAN) is an international system of identifying bank accounts across national borders to facilitate the communication and processing of cross border transactions with a reduced risk of transcription errors. An IBAN uniquely identifies the account of a customer at a financial institution. It was originally adopted by the European Committee for Banking Standards (ECBS) and since 1997 as the international standard ISO 13616 under the International Organization for Standardization (ISO).  

The IBAN consists of up to 34 alphanumeric characters comprising a country code; two check digits and a number that includes the domestic bank account number, branch identifier, and potential routing information. The check digits enable a check of the bank account number to confirm its integrity before submitting a transaction.  

The check digits are calculated using MOD-97-10 as per ISO/IEC 7064:2003`[10]` (abbreviated to mod-97 in this article), which specifies a set of check character systems capable of protecting strings against errors and which are likely to occur when people copy or key data. In particular, the standard states that the following can be detected:  

All single substitution errors (the substitution of a single character for another, for example 4234 for 1234)  
>All or nearly all single (local) transposition errors (the transposition of two single characters, either adjacent or with one character between them, for example 12354 or 12543 for 12345)  
>All or nearly all shift errors (shifts of the whole string to the left or right)  
>High proportion of double substitution errors (two separate single substitution errors in the same string, for example 7234587 for 1234567)  
>High proportion of all other errors
