```python
# Step 3: With robust exception handling

try:
    print("Let's divide two numbers.")
    a = 20
    b = "a"
    print("Result:", a / b)

except ZeroDivisionError:
        print('ZeroDivisionError - You cannot divide by zero, sorry.')

except ValueError:
        print('Value error - You must enter an integer value.')
    
except TypeError:
        print('Type error - You must enter an integer value.')

except:
	    print('General exception - You have raised an exception …. Deary me!')


print('Finish')
```

**Exception Hierarchy**

If we examine the exception hierarchy for the ZeroDivisionError we find it is four levels down:
>Base exception  ->  Exception  ->  ArithmeticError  ->  ZeroDivisionError

According to our rules, if any of the higher exception levels are caught, then the lower ones will not execute.
Looking at the code below and the exception hierarchy, whch exception will catch the zero division error?

Run the code below:


```python
# Step 4: The exception hierarchy

try:
    print("Let's divide two numbers")
    a = 20
    b = 0
    print("Result:", a / b)
	
except BaseException:
    print("Base Exception")

except ArithmeticError:
    print("Arithmetic Error")

except ZeroDivisionError:
	print("Zero Division Error")

print("Finish")
```

You should undertsand that the hierarchy runs top to bottom as expected, if a condition such as a zero division error is to be caught, it will be caught by the first appropriate level in the path to the ZeroDivisionEerror – in this case, the very top level BaseException.  

It is true to say that trying to catch errors at this level is so vague it’s almost pointless, the problem could be any of exceptions in the hiearchy, any of the 66 possible.

Remove the BaseException and its print statement. You should understand that it will be the ArithmeticError that is now caught – run the code and check this.

The lesson here is that if you know what a problem is likely to be, then try and catch it at the appropriate level, you can use the general exception to capture everything else.


**The Exception Object**

**Getting smarter with exceptions**

An exception object in Python is a runtime instance of a class that inherits from BaseException. It encapsulates information about an error that occurred during program execution. You may find this difficult to undertsand as we have not yet studied Object oriented programming. We will, and you can come back to this with confidence. For now, accept that there is some entity that understands and can reproduce the details of an exception.

A more compact version of the tierd hierarchy can be constructed using the following code: 


```python
# Step 5: Using the exception object
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    print("Result:", a / b)
except Exception as e:
    print("Something went wrong:", type(e).__name__)
    print("Details:", e)

print("Carry on")
```

**Custom Exceptions**

It is quite possible to create your own exceptions that provide:   

>More semantic clarity (e.g., TooColdError)   
>More specific control over error handling   
>A way to signal domain-specific problems in your code  
 
Custom exceptions require a little knowledge of class structure and inheritance but for this purpose, please take the following structure as a given and from which you can copy and create your own examples.

All custom exceptions should inherit from Exception (not BaseException, which is reserved for system-level events) thus:


```python
# step 6 – custom exceptions

class TooColdError(Exception):
    pass


# main
temperature = 40
if temperature < 50:
    raise TooColdError("Tea is too cold!")
```

when you run this code, you will find it is not very gracious.    
As we have seen already seen, the use of the try block can moderate the result somewhat:

Lets try this version


```python
# step 7 – custom exceptions

class TooColdError(Exception):
    pass


# main
temperature = 23
try:
    if temperature < 50:
        raise TooColdError("Tea is too cold!")
        
except TooColdError as e:
    print("Warning:", e)
    print("Handle the problem here... then")


print("Continue with the rest of the program ... here")
```

You should now be able to make sense of this custom exception handler and may want to use it.

**Finally**

You may be wondering what has happened to the finally block in our opening paragraphs. Here it is:

```python
# Step 5: With exception handling

try:
    print("Let's divide two numbers.")
    a = 20
    b = "A"
    print("Result:", a / b)

except ZeroDivisionError:
    print("ZeroDivisionError")

except ValueError:
    print("Value error")

except:
    print("General Exception")

else:
    print("Code executes correctly")

finally:
    print("I think that covers everything!")
```
