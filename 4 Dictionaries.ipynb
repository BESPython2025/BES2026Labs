# Data Structures

## Dictionaries

Recall, A **data structure** is a way of organizing, managing and storing data so it can be accessed and modified efficiently.

The **Dictionary** is another data structure; you have allready seen the 'List' and is one of a number of data structures that is supported by Python; some other common examples being:  

>**[Lists]** A collection of non-contiguous elements established as a sequence of nodes  
**(Tuples)** Used to store an ordered collection of elements, but unlike a list, a tuple is immutable, meaning its contents cannot be changed after creation.    
**Arrays:** A collection of items stored at contiguous memory locations - requires imported modules to run such as NumPy  
**Stacks:** A collection following the "Last In, First Out" (LIFO) principle.   
**Queues:** A collection following the "First In, First Out" (FIFO) principle.  
**Trees:** A hierarchical structure where elements (nodes) have parent-child relationships.  
**Graphs:** A set of nodes connected by edges, used to represent relationships.  

Data structures are crucial for optimizing performance in software applications. 

**Dictionaries:**

>Store data in a key-value pair format, making it highly efficient for organizing and retrieving information.  
Recognised by use of curly brackets {&emsp;}  
are mutable i.e. their data can be changed.    
A dictionary is therefore a set of key-value pairs; each key must be unique - it's not possible to have more than one key of the same value.  
A key may be any immutable type of object: it can be a number (integer or float), or even a string, but not a list.  
A dictionary is not a list - a list contains a set of numbered values, while a dictionary holds pairs of values.  
The len() function works for dictionaries, too - it returns the numbers of key-value elements in the dictionary.  
A dictionary is a one-way tool - if you have an English-French dictionary, you can look for French equivalents of English terms, but not vice versa.   
Dictionaries can be nested.   
Dictionary keys are hashed to make look-up extremely fast.

**Listing the contents of a dictionary using .items() function:**

Notice, it doesn't matter what you call the entries in the **for** loop, so long as they match in the print statement:

```python
dict = {"Dog": "Chien", "Horse": "Cheval", "Cat": "Chat"}

for English, French in dict.items():
    print(English, " -> ", French)
print()
    
for x, y in dict.items():
    print(x, " -> ", y)
print()

for key, value in dict.items():
    print(key, " -> ", value)

```

Notice that English - French , x - y and key - value are just labels in the same way that *colour* in colours was a label for the entries in the list colours.

**Listing the contents using .keys() function:**

```python
dictionary = {"cat": "chat", "dog": "chien", "horse": "cheval"}
for key in dictionary.keys():
    #Print the value associated with the key
    print(key, "->", dictionary[key])
```

If you want to print just the values in the dictionary, the french in this case you can use:

**.values() function**

```python
dictionary = {"cat": "chat", "dog": "chien", "horse": "cheval"}
for french in dictionary.values():
    print(french)
```

**Using the sorted() function**

```python
dictionary = {
              "rabbit": "lupin",
              "dog": "chien",
              "horse": "cheval",
              "cat": "chat",
             }

for key in sorted(dictionary.keys()):
    print("sorted values", dictionary[key])

print()
# or to the same effect:

for french in sorted(dictionary.values()):
    print('sorted value',french)

```

Can you spot the differences here and how the sort occurs in each expression? The first sorts keys and the second sorts values, Not easy to spot!

**Changing dictionary entries**

```python
dictionary = {"rabbit": "lupin",
              "dog": "chien",
              "horse": "cheval",
              "cat": "chat",
             }

# Replacing the VALUE for the key ‘cat’
dictionary['cat'] = 'minou'
print(dictionary)

# Adding swan to the dictionary
dictionary['swan'] = 'cygne'
print(dictionary)

# Updating duck to the dictionary
dictionary.update({"duck": "canard"})
print(dictionary)

# Deleting the entry ‘dog’
del dictionary['dog']
print(dictionary)

# Deleting the last entry
dictionary.popitem()
print(dictionary)

# deleting a given entry using pop
dictionary.pop("horse")
print(dictionary)
```
