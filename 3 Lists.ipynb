# Data Structures

## Lists

A **data structure** is a way of organizing, managing and storing data so it can be accessed and modified efficiently.

The **List** is one such data structure. It is one of a number of data structures supported by python; some others common examples being:  

&emsp;&emsp;**{Dictionaries}** Stores data in a key-value pair format, making it highly efficient for organizing and retrieving information  
&emsp;&emsp;**(Tuples)** Used to store an ordered collection of elements, but unlike a list, a tuple is immutable, meaning its contents cannot be changed after creation.    
&emsp;&emsp;**Arrays:** A collection of items stored at contiguous memory locations - requires imported modules to run such as NumPy  
&emsp;&emsp;**Stacks:** A collection following the "Last In, First Out" (LIFO) principle.   
&emsp;&emsp;**Queues:** A collection following the "First In, First Out" (FIFO) principle.  
&emsp;&emsp;**Trees:** A hierarchical structure where elements (nodes) have parent-child relationships.  
&emsp;&emsp;**Graphs:** A set of nodes connected by edges, used to represent relationships.  

Data structures are crucial for optimizing performance in software applications. 

>A list *can be* a collection of non-contiguous elements established as a sequence of nodes, where each node contains data and a reference to the next node in the collection.  
A Python list is actually a *Dynamic Array* - chosen for speed in random access but sacrifices slow insert and delete in the middle of the list to achieve this.   
Each element of a list can be a **scalar** (e.g. integers, floats, strings, booleans) or a **composite structure** (e.g. another list, dictionary, tuple, or even an object).      
Lists are always indexed (numbered) starting from '0'.    
It is used extensively and through the remainder of this course.

**Append and Insert**

The **append function** adds a new value to the end of a list, increasing its length.  
The **insert function** inserts a new value at a given position, thereafter moving all the elements down 1 location

```python
# create an empty list and append 5 elements to it 1-5
my_list = []
for i in range(5):
    my_list.append(i + 1)

print(my_list)

# create an empty list and insert 5 new elements
my_list = []
for i in range(5):
    my_list.insert(0, i + 1)  # insert at position [0] and move everything else up 1 place

print(my_list)

my_list.insert(3, 888)
print(my_list)
```

**The Bubble sort**

The bubble sort is used to arrange a randomly assigned list of numbers into an ascending list of values e.g. [12, 10, 6, 2, 4] -> [2, 4, 6, 10, 12]  

The code for this task is shown below, notice this includes our swap mechanism descibed above:


```python
# list to sort
my_list = [12, 10, 6, 2, 4] 

for j in range(len(my_list) - 1):  					
    for i in range(len(my_list) - 1):
        
        if my_list[i] > my_list[i + 1]:
            #swap 
            my_list[i], my_list[i + 1] = my_list[i + 1], my_list[i]
 				 
print(my_list)
```

**Copying Lists**

Lists can be 'copied' by assigning one list to another e.g.

```python
list_1 = [9,8,7,6]
print(list_1)

list_2 = list_1
print(list_2)

list_2[0] = 2
print("List 1 = ",list_1)
print("List 2 = ",list_2)
```

You should see that the actions above leave both lists changed; this might seem counter intuitive - What is happening?  
When you create a list using list_2 = list_1, python creates a pointer for the new list 2 which points to list 1, essentially making them the same list. Thus changes to list_1 affect list_2 and vice-versa. 
To avoid this, one can make copies and other modifications using 'slicing':

**Slicing**  

Slicing involves copying lists in part or in whole. A slice can easily be recognised by the presene of a colon(:) in the list.
use the code below to determine how slicing is going to work:

```python
# recall the indexing starts at '0'

# Copying a whole list
my_list = [7, 8, 5, 3, 2] 
print(my_list) 
new_list = my_list[:] 
print("New list = ", new_list) 
print()

# Copying a slice of one list to create another
# new_list = my_list[start:stop]

my_list = [7, 8, 5, 3, 2] 
print(my_list) 
new_list = my_list[1:3]
print(new_list)
print()

my_list = [7, 8, 5, 3, 2] 
print(my_list) 
new_list = my_list[:3] 
print("New list = ", new_list) 
print()

my_list = [7, 8, 5, 3, 2] 
print(my_list) 
new_list = my_list[2:] 
print("New list = ",new_list) 
print()

# Slicing using negative indexing
my_list = [7, 8, 5, 3, 2] 
print(my_list) 
new_list = my_list[1:-1]
print(new_list)
print()

# Slice Deleting
my_list = [7, 8, 5, 3, 2] 
print(my_list) 
del my_list[1:3] 
print(my_list)
print()

# delete all the entries in a list
my_list = [7, 8, 5, 3, 2] 
del my_list[:] 
print(my_list)
print()

# Delete a list contents and any referencce to it - it ceases to exist!
my_list = [7, 8, 5, 3, 2] 
del my_list
print(my_list)
```

In all the cases above, except those involving deletions, python creates a new list which can be worked on separately from the first.  
Notice you can't print a list that doesn't exist - an error is thrown!
