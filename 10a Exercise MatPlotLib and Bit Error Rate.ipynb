**Background**  

The prevalence of digital media transmission is, in part, due to the increased ability of systems designers to maximise the number of bits transmitted in the available bandwidth. This has leveraged the development of a wide range of digital modulation schemes which will be briefly described to you in a short lecture. These cover the current methods of Amplitude, Frequency and Phase modulation.   

The bearer, (transmission medium) across which the signal is transmitted, will invariably introduce distortion whilst also being contaminated with noise; this leads to errors in the received data and corruption of the received information. The number of errors received is determined by the Bit Error Rate (BER).   

If the number of bits transmitted is 1000 and 1 bit is received in error, then the error rate is one in one thousand bits or 1/1000. The rules of indices lead to the common notation, BER = 10<sup>-3</sup>. Similarly, one bit in error from a million bits received is 1/10<sup>6</sup> = 10<sup>-6</sup>; two bits in error from a total of one million would be 2 x 10<sup>-6</sup> etc. A BER of 10<sup>-6</sup> is considered good; 10<sup>-10</sup> is almost Error free – actually, Quasi-Error free: the difference being that when the BER is so low, you have to wait so long for an error that for most practical purposes, it never happens. In high speed networks, a greater volume of traffic means data is more likely to become corrupt and lower error rates may be required e.g. 10<sup>-9</sup>. BER is therefore highly system dependent. Many protocols have been developed to overcome data errors such as asking for data to be retransmitted etc but these will not concern us here.  

Bit Error Rate is generally a measure of the success of a transmission system and predicting BER in a noisy environment is an important part of system design - without noise the BER would be zero. You may previously have heard of signal to noise ratios (S/N) that describe radio transmission systems; in digital transmission systems it is the Energy per bit (Eb), not the signal power that is used and this results in a Eb/No ratio (pronounced Ebb-Know) rather than a S/N ratio. The differences are subtle but they essentially come to mean the same expression of quality: A greater Eb hence Signal power means a better chance of detection or lower error rate. Conversely, more noise in the system leads to a lower Eb/No ratio and a higher BER - Low BER is the goal of all systems.  Like its counterpart (S/N), Eb/No is usually measured in dB.

**The complementary error function (erfc)**

The math.erfc(x) function in Python computes the complementary error function of a given number. It’s commonly used in probability, statistics, and engineering, especially in Gaussian distributions and signal processing.  

The error function (erf) measures the probability that a normally distributed random variable falls within a certain range.  
The complementary error function (erfc) is simply 1 - erf(x), meaning it gives the probability that the value falls outside that range.  

It can be used as follows:

import math

x = 1.34  
result = math.erfc(x)  
print(f"erfc({x}) = {result}")  

**Why is it Useful?**

>Signal Processing  - Helps model noise in communication systems  
>Statistics - Used in probability distributions  
>Physics & Engineering - Appears in heat diffusion equations
