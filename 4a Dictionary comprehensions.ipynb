# Dictionary Comprehensions

You have already seen the use of the '**List comprehensions**' in simplifying aspects of list handling. The same ideas can be applied to dictionaries where they form **Dictionary comprehensions**.

recall, a list comprehension had a syntax of the form:

&emsp;&emsp;&emsp;&emsp;**new_list = [new_item  for n in old_list]**  

Dictionary comprehension use the same syntax to generate a new Dictionary:

&emsp;&emsp;&emsp;&emsp;**new_dict = {new_key : new_value for item in list}**  

Notice the dictionary comprehension uses 'List' in its syntax; this is not an error. If a dictionary of key value pairs is to be produced, it can use a list as a starting point.  

This is not however compulsary and a dictionary comprehension can work with a previously defined dictionary:

&emsp;&emsp;&emsp;&emsp;**new_dict = {key_expression: value_expression for item in iterable if condition}**

In Python, an **iterable** is any object that can be looped over (or iterated through) e.g. a list, string, tuple or range.  
if you can loop through it with a for loop, its an iterable!

This will be far more clear after examination of a few examples, heres the first:


```python
student_score = {'Alex': 80, 'Beth': 25, 'Caroline': 82, 'Dave': 37, 'Eleanor': 52, 'Freddie': 79}

# Find those students who passed this exam:
# new_dict = {key_expression: value_expression for item in iterable if condition}
pass_mark = 60
passed_students = {name: score for (name, score) in student_score.items() if score >= pass_mark}
print(passed_students)
```

Lets examine the statement: 

>new_dict = {key_expression: value_expression for item in iterable if condition}  
passed_students = {name: score for (name, score) in student_score.items() if score >= pass_mark}

You should see that the **new_dictionary** called **passed_students** is going to be based on **condition** (if clause) that the student has scored   >= 60 in this test.  

**passed_students = {name:score ....}**  
The result of this comprehension is a dictionary where the keys are student names (name) and the values are their scores (score). Only students who meet the passing condition will be included.  

**for (name, score) in student_score.items()**  
The comprehension iterates over the key-value pairs in the *student_score* dictionary, where *name* represents a student’s name and *score* represents their score. The *items()* method retrieves both keys and values for iteration.  

**if score >= pass_mark**   
The *if* clause filters out students whose scores are below the *pass_mark* (60). Only students who satisfy *score >= pass_mark* will be included in the new dictionary.  

For each **(name : score)** pair that satisfies the condition, the **name** becomes the key, and the **score** becomes the corresponding value in the  dictionary.

Examine the examples in the pane below, they hint at a way through this quite complex subject by first forming the  *for loop* equivalent:


```python
# Example 1
# Create a dictionary of squares
# do this using your standard loops first
squares = {}
for x in range (1,11):
    squares[x] = x**2
print("Loop of squares ",squares)


# Compare to the standard syntax for a dictionary comprehension
# new_dict = {key_expression: value_expression for item in iterable if condition}

#         squares = {x : x**2 for x in range(1,11) if condition}

# since there is no conditional statement required, the expression as it stands will suffice
squares = {x : x**2 for x in range(1,11)}
print("Comprehension squares ",squares)


# If a condition was placed on this dictionary such that only the squares which were below 50 were used, then
# condition becomes "if x**2 < 50"
# This can then be used in the previous expression to give:
squares = {x : x**2 for x in range(1,11) if x**2 < 50}
print("Conditional comprehension squares ",squares)

```

```python
# You should agree that single line dictionary creation is quite impressive!
> squares = {x : x**2 for x in range(1,11) if x**2 < 50}
```

```python
# Example 2
#  Creates a dictionary mapping of names from a LIST to uppercase versions of their names. {name : NAME}
# This uses the upper() method for strings which is currently unknown to us.

names = ["alice", "bob", "charlie"]

# The for loop equivalent:
uppercase_names = {}
for name in names:                              # for each name in the list of names:
    upper_name = name.upper()                   # convert the name to an upper case version using the upper() method
    uppercase_names[name] = upper_name          # Assign the key called name with the value 'upper_name'

print("Uppercase_names: ", uppercase_names)     # print the new dictionary

# Compare to the standard syntax for a dictionary comprehension
# new_dict = {key_expression: value_expression for item in iterable if condition}

uppercase_names = {name: name.upper() for name in names}
print(uppercase_names)
```

You should appreciate how much effort the single line comprehension saves.  

Lets try another:

```python
# Example 3
# Filter the even numbers from the following list and assign them as 'Even' in a dictionary
numbers = [1, 3, 4, 6, 7, 3, 2, 9, 18, 56, 7, 34, 6, 9, 19, 27, 14, 15]
evens = {}
# using a simple for loop:
for num in numbers:
    #test for even
    if num %2 == 0:
        evens[num] = "even"
print(evens)

# Compare to the standard syntax for a dictionary comprehension
# new_dict = {key_expression: value_expression for item in iterable if condition}
evens = {num :"even" for num in numbers if num %2 == 0}
print(evens)

# it is possible to put another condition on this expression using the logical 'and' so we'll have only those number greater than 10
# eg. num > 10
         
evens = {num :"even" for num in numbers if num %2 == 0 and num > 10}
print(evens)
```

```python
# Example 4

# create a dictionary of names such that the value is the first letter of the name {adam : a} etc

names = ["adam", "henry", "chloe", "ada", "alfred", "wendy", "arthur", "alice", "beth"]

#as a for loop
names_dict = {}
for name in names:
    names_dict[name] = name[0]
print("For loop ", names_dict)

# Using a comprehension:
# Compare to the standard syntax for a dictionary comprehension
# new_dict = {key_expression: value_expression for item in iterable if condition}
names_dict = {name : name[0] for name in names}
print("Comprehension ", names_dict)

# It's almost possible to copy directly from the for loop?
# Do you consider comprehensions readable and efficient?
```
