## while Loops

Loops represent another control expression that executes a sequence of code a number of times. They differ from *if* expressions in that:  

&emsp;'if' expressions execute only once  
&emsp;Loops can execute indefinitely  
&emsp;They require a conditional expression in order to terminate    

Without the conditional expression, which **must change** *within* the loop, the code will never exit the loop and will run indefinitely.  
Examine the code below ensuring you read the comments and fully understand its operation:

```python
# Illustrating the operation of a 'while loop'

# A very low and unexpected number
largest_number = -999999999

# input a number to get started 
number = int(input("Enter a number or type -1 to stop: "))

# enter the loop
# while : conditional expression:    Notice the colon after the conditional expression
while number != -1:
# loop body
    # Replace the largest number with the latest number if its bigger than the current value
    # if not, 'Loop'. Exit the loop by inputting '-1'
    if number > largest_number:
    	largest_number = number
    
    number = int(input("Enter a number or type -1 to stop: "))

print("The largest number is:", largest_number)
```


An instruction or set of instructions executed inside the while loop, is called the loop's body.  
If the conditional expression (number != -1) is True (i.e. number is not -1) the body of the loop is executed, if not, the loop is bypasssed.   
The body of the loop should be able to change the conditional expressions value, i.e it asks for a new input number which must be -1 in order to exit the loop.   
When this happens, the conditional expression becomes False (i.e. *is* equal to -1) and the loop exits. 
On exit, the code prints the largest number.

```python
counter = 5
 
while counter != 0: 
	print("Inside the loop.", counter) 
	counter -= 1
 
print("Outside the loop.", counter) 
```

Notice, the conditional expression that determines whether or not the loop is executed is *'count ! = 0'*. Since the count is previously set to 5, the statement is True i.e. the counter is not zero, it is five and the loop executes . 
The loop body will eventually decrement the counter to 0 and the loop will exit. 

**Notice counter = 0 is not the last iteration of the loop, it exits when the counter is 0.**   
Each time the loop executes, it prints the message and the value of the counter.

```python
counter = 5

while counter:
    print("Inside the loop.", counter) 
    counter -= 1
 
print("Outside the loop.", counter) 
```

What is happening?  

Python's while loop condition doesn't need to be an explicit comparison. It evaluates the *truthiness* of the condition directly:  

&emsp;For numbers, 0 is *falsy*, and any non-zero number is *truthy*.  
&emsp;For strings, an empty string ("") is falsy, while any non-empty string is truthy.  
&emsp;For other containers, empty containers, [ ] { } ' ', are falsy, while non-empty ones are truthy.  

Since for 5 iterations of the loop 5, 4, 3, 2, 1, the counter is > 0 and *Truthy*
Thus, *while counter:*  simplifies the code, but it achieves the same behavior as  *while counter != 0:*,  when you're specifically dealing with numbers.  

```python
# Loop comparisons

# Using while counter != 0
counter = 5
while counter != 0:
    print(f"Counter is: {counter}")
    counter -= 1

print("Finished with 'while counter != 0' loop.\n")

# Using while counter
counter = 5
while counter:  # Implicitly checks if counter is 'truthy'
    print(f"Counter is: {counter}")
    counter -= 1

print("Finished with 'while counter' loop.")
```

## for Loops

For loops are particularly useful when you know exactly how many 'turns of the loop' you need to make.  
Examine and run the two code segemnts below:  

```python
# While Loop
i = 0
while i < 10: 
	print(i)
	i += 1
print("While loop Finished")


# For loop
for i in range(10):
    print(i)

print("For loop Finished")
```

Both loops achieve exactly the same thing but the *for* loop does it slightly more efficiently. The count that is explicit in the *while* loop, is now implicit in the *i in range(10)* instruction of the *for* loop.  

You may also appreciate that execution of the while loop is actually optional, if the i counter is set to greater than 100, then the while loop won't execute at all. The *For* loop will always be executed. This my be important in some of your future code.

## Break and continue

**break** and **continue** are instructions that can interrupt the operation of a loop:  

The  **break** instruction in Python is used to immediately **exit a loop**, regardless of whether the loop's condition has been fully met or not. It's typically employed within loops like  or  to interrupt the flow when a certain condition is satisfied.


```python
# Illustrating 'break'

for number in range(10):
    if number == 5:
        break
    print(number)
print("for loop terminated")
```

Notice that when the conditional expression, 'if number == 5' is False, the code proceeds to print the number but when it is True, the code breaks and prints the terminal message. Contrast this with the operation of 'continue':

The **continue** statement in Python is used to skip the rest of the current iteration of a loop and jump directly to the next iteration. Unlike **break**, which exits the loop entirely, **continue**  immediately starts the next iteration, regardless of the remainder of the loop code.

```python
# illustrating 'continue'

for number in range(1, 11):
    if number % 2 == 0:  # Check if the number is even
        # Skip the rest of the loop for even numbers 
        # i.e. don't print the number
        continue  
    print(number)
```

You should see that the code only prints odd numbers, the even numbers are detected and the print() statement is skipped over.
