**Introduction**

Imagine you are a software engineer who has spent a good deal of time determining which classes, methods and variables are needed in a given project. Having done so, you need to convey this information to the coder who will implement your design.

What you dont want to do is implement the methods yourself - that's the coders job. You just need to notify them of the requirement so that their output matches the expected requirements. Their coded output will then not only have the correct structure you specified but it will also work correctly in conjunction with the other parts of the system created by other teams. How do you do this?

>You use as **Abstract class**

What is an abstract class? 

It’s essentially a contract, a way for the designer/architect to say:

>*“Any subclass of this abstract class must implement these behaviours, must expose these attributes and must conform to this shape.”*

It's not about code reuse, inheritace shortcuts ot the improvement of performance, its about guarantees that the coders output will match that of the designers' expectations. It does not dictate *how* things are done but *what* must be done. 

**Example**

**The Stack**
We previously built a stack with two main methods:

>push()  
>pop()  

We also suggested that it could be enhanced by using an additional set of methods such as:

>**peek():**&emsp; Allows you to view the top element without removing it.  
**is_empty():**&emsp;Determines whether the stack has elements or is empty.  
**size():**&emsp;Returns the number of elements in the stack.  
**clear():**&emsp;Removes all elements from the stack.  
**search():**&emsp;Checks whether a particular value exists in the stack.  

For you to convey this requirement to a coder, you create an abstract class that looks like this:

```python
from abc import ABC, abstractmethod

class AbstractStack(ABC):

    @abstractmethod
    def push(self, item):
        """Add an item to the top of the stack."""
        pass

    @abstractmethod
    def pop(self):
        """Remove and return the top item of the stack."""
        pass

    @abstractmethod
    def peek(self):
        """Return the top item without removing it."""
        pass

    @abstractmethod
    def is_empty(self):
        """Return True if the stack has no elements."""
        pass

    @abstractmethod
    def size(self):
        """Return the number of elements in the stack."""
        pass

    @abstractmethod
    def clear(self):
        """Remove all elements from the stack."""
        pass

    @abstractmethod
    def search(self, value):
        """Return True if value exists in the stack."""
        pass

```

This tells the coder which methods are required, their exact names and any parameters they need, thus allowing other teams to write code that can call this class without error. In specifying the content of the abstract class you are specifying the 'interface' to the class i.e. those parts that are exposed to the user - in this example, all of it. Other teams will be expecting to use this exact structure and naming. 

**The @abstractmethod**

You may however be concerned about the decorator function *@abstractmethod* preceding each method, what is it?

Unlike Java or C#, Python doesn’t automatically stop you from instantiating a class that’s meant to be “conceptual”.
So if you write:

```python
class Stack:
    def push(self, item):
        pass
```

Python sees:“a concrete class with a push() method that does nothing” and will happily let someone write:

```python
s = stack()
s.push(9)
```

Even though it make no sense to do so - recall, the pass statement is just a place holder and achieves nothing. 

Prior to Python 3.0, The ABC and @abstractmethod decorator were not available and code of the following type was used instead; notice that to prevent the abstract class being inadvertantly used, an exception was raised:

```python
class Stack:
    def push(self, x):
        raise NotImplementedError
```

Post Python 3.0, it is *expected* that ABC and the decorator function @abstractmethod *are* used.

The *@abstractmethod* tells Python:

>this method must be overridden   
this class cannot be instantiated   
subclasses must implement the behaviour   
this is part of the required shape of the system   

Without the decorator *@abstractmethod*, Python has no way to enforce it. Your abstract class becomes a *normal class* meaning:

>you *can* instantiate it  
you can call its empty methods  
subclasses are not required to override anything  
your architectural contract disappears  
your “shape” becomes optional  

the decorator *@abstractmethod* marks the method as "must be implemented", i.e. overridden. To be used correctly, it is expected that a subclass inherits the structure of the abstract class and overrides **all** its methods.  

When writing an abstract class you musn't mix abstract methods with your ideas of how to implement them - the methods must always be empty (pass) and the whole class is designed to be overridden. The overriding class will then be forced to adopt all the behaviours within it and the designers specification will be fully met.
