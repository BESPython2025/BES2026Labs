**Introduction**

**enumerate()** is a built-in Python function that adds a counter to an iterable. Recall, an iterable is any structure that you can iterate through such as a ListIt[], tuple(), dictionaries{}, string etc. It returns each item as a tuple: (index, value).

A simple example is given below:

```python
# Example 1

planets = ['Mercury', 'Venus', 'Earth', 'Mars']

for i, planet in enumerate(planets):
    print(f"{i}: {planet}")
```

Notice that the *for* line unpacks the tuple created by *enumerate*. It extracts both the index *and* the name of the planet from the list and assigns them to the variables *i* and *planet*.

```python
# Example 2

# Illustrating how to choose the index of the first element

planets = ['Mercury', 'Venus', 'Earth', 'Mars']

for i, planet in enumerate(planets, start=1):    # Try start =  5
    print(f"{i}: {planet}")
```

Notice the printed list starts at 1 not 0 as seen in example 1: this is preferable when human interaction is being created.

Enumerate can be used in comprehensions:

```python
# Example 3

planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']

# Start at 0 by default
planet_order = {i: name for i, name in enumerate(planets)}
print(planet_order)

# add 1 to index
planet_order = {i + 1: name for i, name in enumerate(planets)}
print(planet_order)

# use start as before
planet_order = {i: name for i, name in enumerate(planets, start=1)}
print(planet_order)
```

**Enumerate() with range()**

We have seen how enumerate can become a better replacement for  for …Range() loop. However, they can work together:

Run the code of example 4:

```python
# Example 4

for i, j in enumerate(range(10, 20)):
    print(f"Index {i} -> Value {j}")
    
print("or")

for i, j in enumerate(range(10, 20), start=1):
    print(f"Index {i} -> Value {j}")
```

This can be put to a amore  practical use; imagine allocating half hour interview time slots for a set of students:

```python
# Example 5

names = ['Alice', 'Ben', 'Cara', 'David', 'Ella', 'Frank', 'Grace', 'Hassan', 'Ivy', 'Jack']
hours = range(9, 19)  # 9am to 6pm

for i, hour in enumerate(hours):
    print(f"{hour}:00 -> {names[i]}")
```

You can probably talk yourself out of using enumerate in these examples, they don’t seem powerful enough, exercise 5 was probably easy enough without it but sometimes it really does make a difference; 

consider the following code – it finds the first negative number in the list of *numbers* which must then be displayed along with its index:

```python
# Example 6

numbers = [3, 7, 2, -5, 4, -1]

for i, num in enumerate(numbers):
    if num < 0:
        print(f"First negative number is {num} at index {i}")
        break
```

without enumerate, how could this be done? Index i would have to be tracked e.g. set up a i = 0 count and then increment count every time a test was done, i+=1 etc.
Clearly this is not a critical situation but you must agree, the enumate code is neater and more elegant – more pythonesque!

```python
# Non ennumerate code for example 6

numbers = [5, 7, 2, 5, 4, -1]

index = -1
for i in range(len(numbers)):
    index += 1 
    if numbers[i] < 0:
        print(f"First negative number is {numbers[i]} at index {index}")
        break
```

You should see that this code is slightly longer, is it better - more readable? This really depends on your familiarity with enumerate. Enumerate does give the tighter code but it is more difficult to undertsand for a beginner.
