# ASCII - American Standard Code for Information Interchange

You should have been formally introduced to the ideas behind the ASCII codes in the lecture:

Characters are stored as binary numbers -Including:  

>&emsp;White spaces – space tab etc  
&emsp;Numbers.  
&emsp;Control Characters.  
&emsp;Each character stored as a byte 2^8 = 256 possible characters.  
&emsp;Full 256 symbols is called **Extended ASCII**.  
&emsp;First 128 characters are in general use in western alphabets.  
&emsp;Each character is represented by a **Code point** which is the number allocated against each symbol e.g. 'A' = 65 and 'a' = 97.

A reminder of extended ASCII is given in the image below:
<img src="images\Extended_Ascii_table.png"/> 

*computersciencewiki.org*

Recall the upper 128 characters (128 - 255) represent the alphabets of different nations depending on the *code page* in use.   

For true 'internationalisation', **Unicode**  is used to encode a vast array characters across all languages; Unicode-15, has 149,186 assigned characters - code points, and uses a **variable length encoding** such as UTF-8 (Unicode Transformation Format) to store these. 

<img src="images\Unicode.png"/> 

*Unocde examples*  

UTF-8 has the ability to store 1,114,112 separate code points designed to cover all writing systems, symbols and emojis worldwide. It encodes each symbol as 1, 2, 3 and 4 bytes per character depending on the complexity of the system:   

>1 byte → Basic ASCII characters (A-Z, 0-9, punctuation)  
2 bytes → Extended Latin characters  
3 bytes → Most non-Latin scripts (Greek, Cyrillic and Hebrew)  
4 bytes → Complex symbols (Emojis, rare chines characters, mathematical symbols)  

The **variable length encoding** optimizes storage - the most common codes use the least bytes, and finds wide use in web pages, databases, and modern applications. **Note** UTF-8 is a coding format, **unicode** defnes the code points: 

<img src="images\UTF8.png"/> 
*UTF-8 Examples* 

In this example, to encode U+1F600 in UTF-8, Python converts it into this hex byte sequence:   

> F0 9F 98 80

&emsp;F0 → Indicates "This character needs 4 bytes"  
&emsp;9F 98 and 80 → The remaining three bytes store the actual code point data.    

Python provides built-in functions to encode Unicode characters into UTF-8 bytes and decode them back into readable text.  

Run the code cell below:

```python
# Encode a Unicode string into UTF-8
text = "Hello, world! 😀"
utf8_bytes = text.encode("utf-8")

# Displays byte representation
print(utf8_bytes)  
```

In the print above, the standard characters in hello, world! don't need spelling out, its the 'old school' single ascii byte - H would be 0x48. Only the exotic emoji is spelled out for us in its full hex 4 byte glory. The b at the head indicates its a *bytes* object in python - not a string

UTF-8 preserves all Unicode characters perfectly

```python
# Decode UTF-8 bytes back into readable text
# utf8_bytes is that from the cell above
decoded_text = utf8_bytes.decode("utf-8")
# Displays original text
print(decoded_text)  
```

Since UTF-8 is variable-length, different characters take up different numbers of bytes.

```python
# Print byte lengths of different characters
print(len("A".encode("utf-8")))       # 1 byte (ASCII)
print(len("é".encode("utf-8")))       # 2 bytes (Latin Extended)
print(len("你".encode("utf-8")))      # 3 bytes (Chinese)
print(len("😀".encode("utf-8")))     # 4 bytes (Emoji)
```

**Slicing**   

allows you to extract a range of characters from a string - notice the range is one more than you actually need

```python
# Slices
alpha = "abdefg"
print(alpha[1:3])
print(alpha[3:])
print(alpha[:3])
print(alpha[3:-2]) # negative values start from the RHS of the string -1, -2 NO zero
print(alpha[-3:4])
print(alpha[::2])  # 0 to end step 2
print(alpha[1::2]) # 1 to end step 2
```
