## print()

This is one of python's built in functions; we will spend some time developing functions of our own but for now, realise that they are a separate part of the computer code which is able to:

>**Cause something to happen within the computer**  e.g. send text to the terminal, create a file, draw an image, play a sound, etc  
**Calculate (evaluate) a value**  e.g., the square root of a value or the length of a given text   
**Return a result to the user**  

Functions are easily recognised by their reliance on double brackets. The brackets contain the data to be worked upon by the function; these data are the functions **Arguments** and are passed into the functions code from which the response is derived.

The **Arguments** of a print statement are a data type called **Strings** and represent text rather than numbers. Thus a typical print statement can be written as:


```python
print("Hello world")
```


Notice the string of text to be printed should be within single ' ' or double " " quotes enclosed by the round brackets.
Strings can contain numbers so long as they are not used in calculations but are read, e.g. 'I have 2 cars'

Print statements can be used to provide separation between lines of printed code:

```python
print("My first piece of code")
print()
print("My second")
print("Look at the space in between")
```

Print statements can contain **Escape Characters** which can be used to modify the presentation of the string. A common one being **newline \n**

```python
print("lets put the second part of this \nsentence on a new line")
print()
print("Notice how close it is to the margin \n unlike this.")
print("You should be able to spot why this is and correct it")
```

print() functions can be used with more than one argument using commas to separate them or use none as we have seen

```python
print("A single sentence")
print()
print("Aye, ", "and here are" , "some more")
```

The word 'print' is one of many reserved words that have special meaning within the syntax of the language and can't be used as variable (names) within your code, others that are print related are **end** and **sep**

**end** is a conjunction that determines what joins two print statements, for example

```python
print("My name is", end="*")
print("Monty Python.")
```

Notice that there are no characters in the end= "" string i.e. "". try putting a space or other character between them e.g. end=" "  and see what happens.

The sep="" determines what character or symbol will join two or more strings together

```python
print("This", "is", "my", "world", sep="")
print("This", "is", "my", "world", sep=" ")
print("This", "is", "my", "world", sep="**")
print("This", "is", "my", "world", sep="-")
```

You can combine sep= and end= in a single statement:

```python
print("This", "is", "my", "world", sep="-", end="**")
print("This", "is", "your", "world", sep="_", end="\n")
print("wow!")
```
