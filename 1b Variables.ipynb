# Variables and assignment operators

### What is a **variable**?

A **variable** is a symbolic representation or a named storage location in programming that is used to hold, reference, or manipulate data. It acts as a container for storing values, which can change or remain constant during the execution of a program. Variables enable abstraction and reusability in computations and algorithms.

Each variable must have:  
>&emsp;A unique name - an **identifier**.   
&emsp;Identifiers in Python are case-sensitive.     
&emsp;They must begin with the underscore _ , or a letter, **NOT** a number.  
&emsp;They cannot be a Python keyword e.g print or round etc.   
&emsp;Python etiquette expects the use of **snake_case** - the first letter of a word is always lower case and words are joined using underscores e.g.  my_variable.  

Unlike many other languages, python does not require variables to declared before use - it is **Dynamically Typed**, this means that python can determine how to store the data based on what is written, not what is declared. **Declaration** of a varaible involves stating what type of data it may represent and then assigning it a value.

For example, the C++ language would expect a variable declaration as follows:  

>&emsp;int height = 3

which tells the compiler it needs enough memory to store an integer and assigns that variable the value 3. If you tried height = 3.1, you would get an error as 3.1 is not an integer. Python is more forgiving and would allow reassignment of the new data.

To assign values to a variable we use an **assignment operator** i.e. " **=**" e.g.   **my_variable = 20**

Identifiers which are **constant** e.g. pi = 3.142 etc are written using all caps e.g. PI = 3.142

## Compund operators

are quick ways of assigning variables to themselves. For example, you may be keeping track of a 'count' and wish to inccrement it by 1. You may chooose to do this:

&emsp;count = count + 1

which is acceptable but can be done better:

&emsp;count += 1

This achieves exactly the same as the previous code but a little shorter.

try the code below to see how these operators can be exploited:

```python
a = 5
# Addition assignment
a +=1
print("a =", a)

# Multiplication assignment
a *= 5
print("a =", a)

# Division assignment
a /= 5
print("a =", a)

# subtracting assignment
a -= 1
print("a minus 1 =", a)

# Floor division assignment
a //= 4
print("a floor div =", a)

# Modulus assignment - gives the remainder  of a division
a = 20
a %= 3
print("a modulus 3 =", a)

# Exponentiation assignment 
a = 2
a **= 3
print("a exponent 3 =", a)
```

Other 'bit wise' assignments are possible but are not covered in this module.  

It is worth pointing out at this stage just how changes during the execution of this code - every instruction involved 'a', a changed according to the sequence, when a was multiplied by 5 it became 30, the next instruction assumed a was thirty. Beware that you cant assume a stays at 6, you must always look to the last moment that a was changed - you cant go back in time unless you are willing to redefine (reassign) a to the value you want.

# The input() function

Takes the 'input' from the keyboard and assigns it to a variable e.g. my_message = input()
Whatever is typed at the prompt is then assigned to my_message as a string after **Enter key** is pressed

```python
my_message = input()
print("My message = ", my_message)
```

input() can be accompanied by a message prompt to the user:


```python
my_message = input("Please enter your name: ")
print("Your name is", my_message)
```

Recall the input() function returns a **string**. This can lead to problems, try the code below:


```python
account_balance = 500
deduction = input("How much money do you want to take out?  ")
account_balance -= deduction
print("Your balance is", account_balance)
```

You should find that an error is raised.
This is because we're trying to use the input string as a number!

The way to convert your string input to a number, either integer or float, is to use 'Type casting'.

# Type casting

Type casting converts one variable type, string, float or int, to another, recall:

>&emsp;Integers = whole number e.g. 1678 -235 100023  
&emsp;Floats or 'floating point numbers' have a fractional part displayed as a decimal e.g. -35.67, 1.32 or 30456.56  
&emsp;Strings are always textual information contained in within quotation marks e.g. "This is a message" or "My name is.."  

Thus:  
&emsp;int() causes any string within the bracket to be converted to an integer.
similarly:  
&emsp;float() converts strings to floating point numbers  
also:  
&emsp;numbers can be converted to strings so they may be treated as such in the appropriate parts of your program using:  
&emsp;str()

Try the code below:


```python
# account balance should now be deducted correctly
# as your string input is converted to a float
account_balance = 500
deduction = float(input("How much money do you want to take out?  "))
account_balance -= deduction
print("Your balance is", account_balance)

# 4 is a string which is then converted to an integer
x = int("4")
y = x**2
print("x squared = ", y)

# The number 4 is converted to a string for printing
x = 4
my_string = str(x)
print("This is the number 4:   " + my_string)
```

# String concatenation

It is useful to be able to join strings together to create a composite. This process is known as **concatenation** and in python is achieved using the '+' operator as seen in the last example above:


```python
first_name = input('What is your first name?')
surname = input('What is your surname?')
address = "mod.gov.uk"
url = first_name + "." + surname + "@" + address
print(url)
```

Only strings can be concatenated; trying to incorporate a number will fail

Concatenation can be used with string replication to create simple graphics:

# String replication

strings can be replicated any number of times using '*'

Thus:  
>"Name" * 3 &emsp;becomes &emsp;NameNameName  
3 * "an" &emsp;becomes &emsp;ananan  
5 * "2" &emsp;becomes &emsp;22222

```python
triple_name = "Name" * 3
print(triple_name)

an_some = 3 * "an" 
print(an_some)

treble_two = 3 * "2"
print(treble_two)

#Simple graphics
print("+" + 10 * "-" + "+") 
print(("|" + " " * 10 + "|\n") * 5, end="") 
print("+" + 10 * "-" + "+") 
```

# f strings

f strings represent a very useful and simple way to incorporate both text and data in one statement:

```python
x = 3
y = 2

result = x + y

print(f"Adding x and y = {result}")
```

Notice how the result within the curly bracket is reproduced from the earlier code. This avoids using type casting of results prior to concatenation.
In fairness, the following code also works


```python
x = 3
y = 2

result = x + y

print(f"Adding x and y = ", result)
```

But the real beauty of f strings is that many types of expression can be carried out within the { } brackets; try predict the out of the following expressions before running the code:

```python
#Performing calculations
x = 3
y = 2
print(f"Adding x and y = {x + y}")

# Invoking functions
name = "Bob"
message = f"{name.lower()} is written in all lowercase now."
print(message)

#Incorporating formating
large_number = 1000000
growth_rate = 0.075
message = f"The company earned £{large_number:,} last year and grew by {growth_rate:.1%}."
print(message)

#Combining multiple variables and expressions
quantity = 5
item = "apples"
cost_per_item = 0.50
message = f"You bought {quantity} {item} for a total cost of £{quantity * cost_per_item:.2f}."
print(message)
```

f strings can really simplify code.  

Though we haven't covered this topic yet, conditional logic can also be incorporated, again try and predict the out of the code before you run it:



```python
score = 85
grade = f"You {'passed' if score >= 60 else 'failed'} the test with a score of {score}%."
print(grade)
```

### There are a number of other types of notation and conversion that can be used:

&emsp;Scientific (E or e)  
&emsp;percentage (%)  
&emsp;Binary (b)  
&emsp;Hexadecimal (x or X)  
&emsp;Octal (o)  


```python
# Formatting in scientific notation

x = 123456
print (f"{x:.2e}")
print (f"{x:.2E}")

# formatting using %
number = 0.2345
print (f"{number:.2%}")

# different kind of format
number = 5678912345
formatted_number = f"{number:.2e}"
print(formatted_number)

# Binary
number = 53 
# precision not allowed
print (f"{number:b}")
```
