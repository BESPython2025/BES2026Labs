**Introduction** 

What Is Polymorphism?  

Polymorphism allows objects of different classes to be treated as if they were objects of the same class.

This means methods can be overridden or reused in a flexible way, enabling code to handle different types without knowing their exact class in advance.

Types of Polymorphism -There are two main types:  

>Method Overriding (Runtime  - Dynamic - Polymorphism): A subclass provides a specific implementation of a method from a parent class.   
>Method Overloading (Compile-time - static - Polymorphism): The same method name is used with different parameters in the same class - Python does not support true method overloading, but can achieve similar behavior. We shall ignore it here.


**Method Overriding (Runtime Polymorphism)**

This allows subclasses to redefine methods from their superclass.

Examine the code below:

There are two classes, named One and Two, Two is derived (inherits) from One - Nothing special there. However, the do_it() method is declared twice, once in each method - originally inside One and subsequently inside Two.  
The essence of the example lies in the fact that it is invoked from inside class One using the **doanything()** method.  

Try and predict the output and then run the code:


```python
#Illustrating Polymorphism

class One:
    def do_it(self):
        print("do_it from One")

    def doanything(self):
        self.do_it()
        
class Two(One):
    def do_it(self):
        print("do_it from Two")

#main
one = One()
two = Two()

one.doanything()
two.doanything()
```

The first invocation is straight forward, invoking **doanything()** from the object named **one** invokes its own method **do_it** which will then successfully print the message **do_it** from one

The second invocation needs some attention. It's relatively simple too if one can remember the **Method Resolution Order (MRO)** i.e. the order in which python finds methods when overriding- Recall this follows the pattern **Own class, then Upwards and Left to right**, Since there is no multiple inheritance in this scheme, we can ignore the left to right.   
The second invocation will launch do_it() in the form existing inside  class **Two** , regardless of the fact that the same invocation takes place within the One class.

In effect, the code produces the following output:

>do_it from One   
>do_it from Two  

Note: the situation in which the subclass is able to modify its superclass behavior (just like the example) is called polymorphism. The word comes from Greek (polys: "many, much" and morphe, "form, shape"), which means that one and the same class can take various forms depending on the redefinitions done by any of its subclasses.

A **virtual method** is a method that is *defined in a superclass* but can be redefined (overridden) in a subclass to alter its behavior. This ensures that when an object of a subclass is used, it calls the overridden version instead of the superclass's version. Declaring a method as virtual means: "Feel free to replace me in a subclass if needed."

In other languages, C# for example, methods need to be declared as *virtual* e.g

>class Animal {   
    &emsp;public virtual void Speak() {   
        &emsp;&emsp;Console.WriteLine("Some animal sound");   
    &emsp;}   
}   

Whereas, in Python, all methods are virtual by default. You don’t need to declare them specifically. That means any subclass can override a method from its superclass unless it's intentionally blocked (e.g., with name mangling or design choices).


**Why Are Virtual Methods Useful?**

>Enhance flexibility: Allows different behaviors in subclasses.  
>Promotes code reusability: Superclass provides a default, and subclasses refine it.  
>Supports dynamic method calls: Decides behavior at runtime, not just compile time.  

You may feel that this way of executing methods could be achieved in a less roundabout way and you would be correct but the example is there to illustarte the bigger picture.....

**Why Is This Useful?**  

>**Encapsulation of Behavior:**&emsp;The doanything() method ensures all subclasses retain a standardized way to trigger do_it(), without needing to rewrite logic for every new class.   
**Polymorphism in Action:**&emsp;Even though doanything() is only defined in One, it automatically calls the overridden version of do_it() in Two—this showcases dynamic method resolution!   
**Extensibility:**&emsp;if you later add a Three class that overrides do_it(), you don’t need to modify doanything()—it will always call the correct version automatically.  
**Code Reusability:**&emsp;Avoids duplicate logic across classes. If you had multiple subclasses, rewriting do_it_two(), do_it_three(), etc., would be messy.  

Lets examine another more practical example:

```python
# Using polymorphism
# Notification has a 'virtual method' - send() overridden
# by all the subclasses below

class Notification:
    def send(self, message):
        print("Generic notification:", message)  # Default behavior

class SMSNotification(Notification):
    def send(self, message):
        print(f"Sending SMS: {message}")

class EmailNotification(Notification):
    def send(self, message):
        print(f"Sending Email: {message}")

class PushNotification(Notification):
    def send(self, message):
        print(f"Sending Push Notification: {message}")


notifications = [SMSNotification(), EmailNotification(), PushNotification()]
for notifier in notifications:
    notifier.send("System update available!")
```

This is an example of runtime polymorphism -  where the method call is dynamically resolved based on the actual object type at runtime, rather than at compile time. 

Notice that each subclass SMSNotification etc, inherits from the superclass, Notification, and since each subclass has a method send() which conflicts with the superclass method of the same name, the order of inheritance has to determine which of the send() methods to use, the inheritance MRO would favour the subclass since the search is own, bottom to top (excluding L to R in this case- no multiple inheritance).  

Allowing each subclass to share a common method (interface), send() in this example, makes the interface easy to implement for all subclasses and any that might be built in future. The presence of send() in the superclass forces the MRO to do choose the right one.


On reading this code you may be tempted to think this is all very obvious, you can even run it without the Notification class - comment it out and try and run it!

**Where Polymorphism Comes In**

Instead of manually checking what type of notification we’re dealing with (if email, do X... if SMS, do Y... etc), we simply call send()—and Python automatically invokes the correct version at runtime.

lets expand on this statement. If we didn't have inheritance and polymorphism, our loop:

**Why This Matters in a Real App**

Python doesn’t need inheritance for polymorphism — it only needs objects with the same method name.
The base class is there to define the interface, not to make the polymorphism work.

This technique allows flexibility. If tomorrow we add WhatsAppNotification, we don’t need to modify existing code—it just works because everything obeys the same interface.


There is another type of polymorphism in python called **Duck typing**, it is runtime polymorphism *without* inheritance

```python
# Behavioral Polymorphism - duck typing

class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

# No inheritance, but both objects behave similarly
animals = [Dog(), Cat()]
for animal in animals:
    print(animal.speak())  # Calls the correct method dynamically
```

Here there is no inheritance being used and no superclss in sight.   
Here we're relying on 'behaviour consistency' instead.  

**Behavior consistency** refers to the idea that objects in a system can be used interchangeably if they provide the expected behavior—regardless of their actual type. This is a key aspect of polymorphism and is especially important in duck typing, which Python heavily relies on. 

**What Does Behavior Consistency Mean?**  

>Objects don’t have to share the same parent class—as long as they implement the same methods and behave similarly.  
>Code becomes more flexible—because it focuses on what an object can do, rather than what type it is.   
>Avoids unnecessary type checking—so functions can work with different objects, even if they belong to separate hierarchies.  

Consider the example below:


```python
# Duck type polymorphism

class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Robot:
    def speak(self):
        return "Beep Boop!"

# Function that expects behavior consistency - duck typing
def make_noise(entity):
    return entity.speak()  # No need to check type

# Works for all objects because they all have a speak() method - behaviour consistency
print(make_noise(Dog()))    # Output: Woof!
print(make_noise(Cat()))    # Output: Meow!
print(make_noise(Robot()))  # Output: Beep Boop!
```

**Why is 'Behavior Consistency' useful?**  

>**Scalability:**  &emsp;Future objects (e.g., Parrot, AI Assistant) can be added as long as they implement speak().   
>**No need for type checking:**  &emsp;We never ask "Is this a Dog or a Cat?"—we just call speak().   
>**Encapsulates behavior:**  &emsp;The system focuses on functionality rather than rigid class structures.  


**When Does It Become Polymorphism?**  

If we had a common superclass (e.g., class Animal: with a speak() method), then we’d clearly be demonstrating runtime polymorphism, where method resolution happens dynamically at runtime based on the object's actual type.
But with duck typing, the term “polymorphism” can feel like a stretch because we’re not enforcing a hierarchical structure—just ensuring method consistency across objects - It’s more about adaptability than strict polymorphism.

You may feel that Polymorphism is just glorified Inheritance; you wouldn't be far wrong but it is important to undertsand the differences:

>Inheritance is about structure: a subclass receives attributes and methods from a superclass. But that alone isn’t polymorphism.

Polymorphism kicks in when 

>The subclass overrides a method from the superclass   
>Code refers to objects via the superclass type, but the actual method behavior depends on the subclass  

So, inheritance becomes polymorphism when the subclass not only inherits but customises behavior, and we interact with it using a reference to the base type.

>Inheritance ≠ polymorphism  
>Inheritance + overriding + base-type interaction = runtime polymorphism
