```python
# Experiment 6 - getting access to superclass 'instance' variables

class Super:
    sup_classvar = 100

    # declare an instance variable supvar
    def __init__(self):
        self.supvar = 11     
        
    def printme(self):
        return ("This message")

    
class Sub(Super):
    sub_classvar = 101

    def __init__(self,sub_name):
        super().__init__()      # calls the initialiser of the superclass
        self.name = sub_name

    def printmetoo(self):
        return ("My name is "+ self.name)

    def get_name(self):
        return(self.name)
    

#Main
#create an instances of the class
sub_obj =Sub('Dickens, Charles Dickens')

#can we access the methods of the sub and sup classes?
print(sub_obj.printme())
print(sub_obj.printmetoo())

#Do we still have access to the class variables 
#of each class from the sub class
print (sub_obj.sub_classvar)
print (sub_obj.sup_classvar)

#get the name associated with the class
print(sub_obj.get_name())

#Access the superclass variable supVar = 11
print(sub_obj.supvar)		# Raises Attribute error
```

The super().`__init__()` call is a crucial part of inheritance in Python. It allows the subclass (Sub) to initialize the attributes and methods of its parent class (Super), ensuring that the inherited functionality is properly set up.

Without the super().`__init__() code in the subclass, this code, this code would not instansiate  the class Super - unlike experiment 3; so its constructor is never executed and neither is the *self.supvar = 11* instruction - that is until we *force* its constructor to execute from the subclass.   

In this way, the subclass get full access to the superclass instance variable.

If the super class was never going to be instantiated but it was a useful class with useful methods that could be used elsewhere in our code, then it mght be better to send the variable e.g. supavr = 11 to the superclass; see the code below.

```python
class Super:  
    def __init__(self, super_value):  
        self.super_value = super_value  

class Sub(Super):  
    def __init__(self, sub_name, super_value):  
        # Pass argument to Super's constructor  
        super().__init__(super_value)    
        self.name = sub_name
```

## Multiple Inheritance

*Multiple inheritance* is when a class inherits from more than one parent class. This allows a subclass to inherit attributes and methods from multiple sources.  
Examine the code below, particularly note how the empty class C inherits from both A and B using the class declaration: **class C(A, B):**   
Predict the codes ouptut and then run it:


```python
# Experiment 8

class A:
    def method_A(self):
        return "Method from A"

class B:
    def method_B(self):
        return "Method from B"

class C(A, B):  # Inheriting from both A and B
    pass

obj = C()
print(obj.method_A())
print(obj.method_B())
```

You may be concerned that this might be a pointless, after all we can make B inherit from A by including: Class B(A) .... so what is the point?   
There are times when it is better to keep some classes separate as there is no justification to combine them and create unnecessary relationships.  
Take the follwing code as an example:

```python
class Flyer:
    def fly(self):
        return "I can fly!"

class Swimmer:
    def swim(self):
        return "I can swim!"

class Duck(Flyer, Swimmer):  # Duck inherits from both - fish wouldn't!
    pass
```

Here, Flyer and Swimmer are independent traits, so keeping them separate allows Duck to combine both, without enforcing a hierarchy between them we avoid **deep inheritance chains** and **unecessary dependencies**

## Method Overriding

**Overriding happens when a subclass provides a new implementation for a method that is already defined in its parent class.**
This allows subclasses to modify or extend behavior while maintaining the original method structure.   
In the code below there are three classes, a child class which inherits from its parent class which then inherits from a grandparent class. All these however use a function called show().   
The problem arises when, after creating an object of classchild, we then try and invoke obj.show(), the interpreter has to determine which of the three show() functions to exectue - child, parent, grandparent or all three?

Run the code and find out?


```python
# Experiment 9 
# Overriding

class Grandparent:
    def show(self):
        return "Grandparent"

class Parent(Grandparent):
    def show(self):
        return "Parent"

class Child(Parent):  # Inherits from Parent, which inherits from Grandparent
    def show(self):
       return "Child"

obj = Child()
print(obj.show())  # Output: "Child"
```

It is clear that the child class takes precedence and we say that the parent function has been **overridden**.  

In single inheritance, i.e. only ever inheriting from one source, the child class always takes precedence over the parent when overriding methods - this should make sense because if we didn't want to overide the parent class then we wouldn't have stated our intent by re-writing the function in the child class.
When an object calls a method, Python first looks in the child class — if the method is to be overridden, the child’s version is used. 

>Overriding allows python to: Customize behavior: subclasses can modify inherited methods.

The way that python decides on which overridden functions to run is determined by its **Method Resolution Order** (MRO).


## Method Resolution Order

MRO defines the order in which Python searches for a method in an inheritance hierarchy.  
Lets see how this works. In the code below there are three classes, each with a function show(). After instantiating obj from class D, the problem, as we have already stated, is which show() function should be executed?

Examine the code before running:

```python
# Illustrating  method resolution order (MRO)

class A:
    def show(self):
        return "A"

class B(A):
    def show(self):
        return "B"

class C(A):
    def show(self):
        return "C"

class D(C, B):  # Multiple inheritance
    pass

obj = D()
print(obj.show())  # Resolves using MRO
print(D.__mro__)   # Displays method resolution order
```

In this code, the empty class D, inherits from both B and C and does not attempt to override the show() function of its parent classes - **Direct parent** B and C or **Indirect Parent (Ancestor)** A.  
The order in which python will search for the correct show() function to execute is decided as follows:

> Child class first   
> Classes from bottom to top in the hierarchy in the order of inheritance   
> Left to right when dealing with multiple inheritance   

Which means, given the code above, it will first try D then B (left) then C (right) finally A. It will however find a show() function at B which it will then execute - the other show() functions are ignored.

Lets now address one of the problems that the MRO was designed to solve. It occurs in multiple inheritance situations and is called - the Diamond Problem.  

**The Diamond Problem**

The diamond problem occurs when a class inherits from two parent classes, which themselves inherit from a common ancestor. This is shown i nthe figure below: 

<img src="images\The_Diamond_Problem.png" width="400" height="300">

This creates an ambiguous inheritance structure, where Python must decide which method to call when a method name appears in multiple places. Can you see the problem that might arise if the rules of MRO had not been fully devised?

The problem is: when searching for a method, does the seacrh go - D to B to A or D to B and then to C and finally to A.

Without the rules of the MRO, this is actually a tricky problem and it might be that a method is missed in an unvisited class. Fortunately the rules discussed above clear this up:

> Child class first   
> Classes from bottom to top in the hierarchy in the order of inheritance   
> Left to right when dealing with multiple inheritance

So the search prioritises classes at the same level - L to R before moving up. Thus the order should be: D -> B -> C -> A

**Composition vs Inheritance**

Inheritance isn’t always the best tool—sometimes composition (HAS-A relationships) is better. Consider the following code:

```python
class Engine:
    def start(self):
        return "Engine starting..."

class Car:
    def __init__(self):
        self.engine = Engine()  # Composition instead of inheritance

    def drive(self):
        return self.engine.start() + " Car is driving!"

obj = Car()
print(obj.drive())  # "Engine starting... Car is driving!"
```

The Car and Engine example demonstrates composition because instead of making Car inherit from Engine, we *give* Car an instance of Engine inside it — meaning Car "has" an Engine rather than "is" an Engine. In this case car does not inherit from Engine, as that would imply car is derived from engine in the same way that domestic dog is derived from wolf - clearly not a good relationship for the car. By allowing car to instantiate an engine, the car has an engine rater than being one:

>Car does not inherit from Engine, but instead contains an Engine object as an attribute (self.engine).   
>Car delegates the start() functionality to Engine rather than having it built into itself.   
>This follows a HAS-A relationship rather than an IS-A relationship.  

We can explore these relationships briefly before moving on:

**HAS-A vs IS-A** with Vehicles and motion 

Imagine you’re designing a system for vehicles and their motion. We may be tempted to put all these common attributes in to the base class from which any vehicle can then inherit; this is shown briefly in the code below.

**Inheritance (IS-A Relationship)**

```python
# an IS-A relationship

class Vehicle:
    def move(self):
        return "This vehicle moves."

class Car(Vehicle):  # Car inherits directly from vehicle, Car IS-A vehicle
    def move(self):
        return "The car drives on roads."

class Boat(Vehicle):  # Boat IS-A vehicle
    def move(self):
        return "The boat sails on water."
```

Contrast this with the HAS-A relationship:

```python
class Movement:
    def __init__(self, method):
        self.method = method  # Defines movement type

    def move(self):
        return f"The vehicle moves by {self.method}."

class Vehicle:
    def __init__(self, movement):  # Vehicle comes with (HAS-A) Movement
        self.movement = movement

    def move(self):
        return self.movement.move()  # Delegating behavior to Movement

# Creating movement types for Car and Boat
road_movement = Movement("driving on roads")
water_movement = Movement("sailing on water")

# Assigning movement behavior dynamically
car = Vehicle(road_movement)
boat = Vehicle(water_movement)

print(car.move())  # Output: "The vehicle moves by driving on roads."
print(boat.move())  # Output: "The vehicle moves by sailing on water."
```

Why Is This Composition?

>Vehicle does not inherit movement behavior—it contains a Movement object.  
>Vehicle delegates movement to Movement, making its logic separate and reusable.  
>Easily swap or modify movement without changing Vehicle itself.

**Why you should use an abstract class**  

Prevents Code Duplication: If every subclass has name and age, an abstract class ensures centralised management.   
Standardises Behavior: Guarantees all subclasses follow a common structure, enforcing consistency.   
Encapsulation: Keeps core logic within the parent class, reducing redundancy in subclasses.  

**When NOT to Use One?**

If There’s No Shared Behavior: If subclasses don’t need a shared implementation, inheritance may be unnecessary.  
If Only One Subclass Exist: In cases where inheritance isn't adding value, composition might be better.  
If Flexibility is Required: Abstract classes enforce a rigid structure, which might be limiting for future extensions.  

**Best Practice:** 

Use an abstract class for conceptual categories, like Person, when multiple subclasses inherit common behavior. However, for small, isolated cases, sometimes simple class duplication is cleaner than unnecessary abstraction.
