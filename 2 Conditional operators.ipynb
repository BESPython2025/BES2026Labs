# Control Statements

There are a number of control statements within Python that give the language its 'decision making' capabilities. These statements are called **control statements**

They are associated with **conditional operators** such as **greater** or **less than** that form the decision.

In English we may choose to say:  

&emsp;"If the weather is good, then i will go for a walk"

Here the **Control statement** is **if** and the **Conditional Operator** is **Equal**

This can be reformulated as:

&emsp;"**If** the weather **equals** good, **then** i will go for a walk"

In a program, this kind of conditional control may translate to the following:


```python
good_weather = True

if good_weather == True:
    #then
    print("Go for a walk")
else:
    print("Stay at home")

print("That was pleasant")
```

You should see that the decision is based on the condtion of the **Boolean** variable **good_weather** which is set to **True**.  
The **if** statement checks the condition of **good_weather** and deciding that it is indeed **True** i.e the weather **is** good, proceeds to print the action "Go for a walk"  
Had good_weather been **False**, implying the weather is poor, then the **else statement** would have been executed instead and i would "Stay at Home"

Change **good_weather** to **False** and check that the altenative **else** code is executed.  

Notice that once the 'decisions' have been made using the **if..else** code block, the code reverts to executing those statements outside the block which in this case is the final print().  
Make sure the code that is part of 'else' is indented otherwise it will be executed regardless. You can't make this mistake in the 'if' becasue it will throw an error. 

It should be pointed out that any number of instructions can be placed after the if and else statements; so long as they are aligned at the indent, they will be executed, thus


```python
good_weather = True
number_of_walks = 5
books_to_read = 3

# If the weather is good do this
if good_weather == True:
    print("Go for a walk")
    print("and see the sights")
    number_of_walks +=1

# if the weather is bad do this instead    
else:
    print("Stay at home")
    print("and read a book")
    books_to_read -=1

# Do this regardless of the previous code
print("Number of walks taken is:", number_of_walks)
print("Number of books to read is:", books_to_read)
```

Notice that any code not aligned with an **else indent** is not part of the if..else block and will be executed regardless of any previous decisions.  
How can you get to read a book and reduce the book count?  
If..else blocks can be **nested** ...


```python
# Illustrating nested if statements

good_weather = False
number_of_walks = 5
books_to_read = 1

if good_weather == True:
    print("Go for a walk")
    print("and see the sights")
    number_of_walks +=1
else:
    print("Stay at home")
    print("and read a book")
    books_to_read -=1
    
    # I've read my book, check my library...?
    if books_to_read == 0:
        print("Buy more books")
        # get more books!
        books_to_read = 10


print("Number of books to read is:", books_to_read)
```

In this example we force the code to take the else block by setting good_weather to False. This makes us read a book after which the *books_to_read* is decremented by 1. This in turn forces the execution of the second if statement as it is aligned correctly in the else block- here we buy new books becasue we were previously down to 1.

## Logical Operators

It is possible to extend the complexity of the decision process using the **Logical operators** **and** and **or** to test a value is within a given range. This would otherwise require a nested if statement. In the example below, if the temperature "is greater or equal to 30 **and**   less than or equal to 70" then the "Temperature is in range"


```python
Temp = 50

if Temp >= 30 and Temp <= 70:
    print("Temperature is within range.")
else:
    print("Temperature is out of range.")
```

The logical **or** operator can be used to the same effect:

```python
Temp = 80

if Temp < 30 or Temp > 70:
    print("Temperature is out of range.")  # notice the temp statements are reversed here to align with the logic
else:
    print("Temperature is within range.")
```

It is also quite possible to use the logical operator **not**:

```python
Temp = 50

if not (30 <= Temp <= 70):
    print("Temperature is out of range.")
else:
    print("Temperature is within range.")
```

Further we could just use:

```python
Temp = 50

if (30 <= Temp <= 70):
    print("Temperature is within range.") 
else:
    print("Temperature is out of range.")
```

There are clearly many options when considering which conditional operators to use. What may be correct and apparent to you may not always appear logical to another.
