**`__init__`**

When an instance of a class is created, a special part of its code, the **class constructor**, defined by **`__init__()`**, is called and executed. Since we have not defined what should happen at this time - we haven't specified the constructor, and nothing happens. However, if we define the constructor, as in the code below, you will see a completely different behaviour to the code given above:

```python
class TheSimplestClass:
    def __init__(self):
        print("My first class")

# Creating three instances
my_first_class = TheSimplestClass()
my_second_class = TheSimplestClass()
my_third_class = TheSimplestClass()
```

What do you notice?
Now, each time the class is instantiated, it prints the message, as a  result of the execution of the constructor; everything below the constructor and indented, is executed on intantiation - again not too useful at this stage but we are getting there.

You should also have noticed the word '**self**' was included in the code above:  
**self** is a reference to the instance of the class that is being created.
It represents the current object; Whenever you create an instance of TheSimplestClass, self refers to that specific object.
If you add attributes to the class, self lets each object store its own unique values.

You will find that when using functions within a class (methods), the use of 'self' is Mandatory. In Python, *self* must be included as the first parameter in any method that belongs to a class instance. We shall see this later.

Lets create and instatiate a more useful class:


```python
# illustrating a more useful Car class
class Car:
    # Class variable (shared by all instances)
    wheels = 4  
    def __init__(self, brand, model):
        # instance variable unique to this instance
        self.brand = brand   
        self.model = model

# main
# Instantiating the Car class multiple times
car1 = Car("Toyota", "Corolla")  
car2 = Car("Ford", "Mustang")  
car3 = Car("Ford", "Focus")  

# acessing the data from each class
print(car1.brand, car1.model)  # Output: Toyota Corolla
print(car2.brand, car2.model)  # Output: Ford Mustang
print(car3.brand, car3.model)  # Output: Ford Focus

# illustrating the use of 'class' variables
print("car1 has", car1.wheels,"wheels")  # Output: 4
print("car2 has", car2.wheels, "wheels")  # Output: 4
```

>**class Car:**   defines the class name

>**def __init__(self, brand, model):**    is the constructor followed by the two parameters passed in on instantiation - brand and model - note the *colon* and *def* statement, these are the variables that are expected to change on each new instatiation - i.e. each object.  
>**self.brand = brand** and  
>**self.model = model**   
>tells us that every separate instance of **car** will have two parameters, brand and model.  
>The self ensures they stay uniques to that instance and not shared across instances. Since these are unique to each instance, they are call **Instance variables**.  Every object (instance) will be created with a unique set of variables that describe it. 
>**wheels = 4**  In contrast to an instance variable, a **Class variable**  is shared by **all** instances

Notice the method of accessing each of the variables using **Dot** notation. This is the same as accessing the functions from within a module.   

When we instantiate a class, we pass arguments to the class constructor (`__init__`), which then assigns them as **instance** attributes. These arguments are received as parameters within `__init__`, and self ensures each object gets its own unique values.


**Class and Instance variables**

Lets take a moment to consolidate the difference between a class and instance variable.  
Examine the following code and then run it:

```python
class MyClass:
    MyClass_var = 101  # Class variable - same for every class

# Creating instances
obj1 = MyClass
obj2 = MyClass

print(obj1.MyClass_var)  # 101
print(obj2.MyClass_var)  # 101

# Changing the class variable via the class
MyClass.MyClass_var = 202

print(obj1.MyClass_var)  # 202 (changed for all instances!)
print(obj2.MyClass_var)  # 202
```

We can conclude that a class variable applies to every instance of that class.  
In the first case we see obj1 and 2 have the same MyClass_Var which is equal to 101.  
In the second case, we alter the MyClass_Var to 202 and once again, it is shared across all instances of the class - this is the behviour of a class variable.  
Instance variables are defined as unique for each object at the moment of instantiation, in the car class above:

        &emsp;&emsp;&emsp;&emsp;self.brand = brand     
        &emsp;&emsp;&emsp;&emsp;self.model = model  

brand and model are the instance variables as they differ from object to object.

In some applications it is likely that some variables, such as names, locations etc may need to be changed - just married!. It is not really appropriate to re-enter the whole of the data, better to provide functions within the class to do this for us.

Examine the following game code, it follows the personnel type of code we have been using but it also provides what is known as a **setter** function:


```python
class GameCharacter:
    def __init__(self, name, species, level, home_base):
        self.name = name
        self.species = species
        self.level = level
        self._home_base = home_base  # Conventionally private

    #################################################################
    # Setter with validation
    def set_home_base(self, new_base):
        valid_bases = ["Skyhold", "Driftwood", "Ironforge", "Rapture"]
        if new_base in valid_bases:
            self._home_base = new_base
        else:
            print(f"Invalid base '{new_base}'. Please choose an existing base.")

    #################################################################
    # Getter method
    def get_home_base(self):
        return self._home_base

    def print_profile(self):
        print(f"{self.name} the {self.species}, Level {self.level}")
        print(f"Base of operations: {self.get_home_base()}")
        #print()


# Create a character
hero = GameCharacter("Kaela", "Elf", 12, "Skyhold")

# Display profile
hero.print_profile()

# Reassign to a valid base - from Skyhold to Driftwood
hero.set_home_base("Driftwood")
print()

# Confirm current base
print("Current base:", hero.get_home_base())
print()

# Try assigning an invalid base
hero.set_home_base("Atlantis")
print()

# Display profile
hero.print_profile()
```

This code has quite a bit going on:  
>The observation we wish you to make is incorporated between the lines of hashes - the setter.  
Notice first that the variable home_base name has been changed to _home_base i.e. underscore home_base. This is not absolutely essential but it introduces a modicum of priviacy into the variable and hints to users that this variable should be set and read via getters and setters.  
The **setter** is a small function that can set the variable *_home_base* when called and can also check the validity of the new entry against a list of items using the keyword **in** - which is sensible .  
It physically does not stop modification by external code but it should help signal that direct access should be avoided in favour of controlled methods like setters.

For example, the following code will also set the home_base name:

but you should understand that the privacy underscore should desuade any *responsible* coder for doing so in favour of the setter provided.

You should also spot another function called **get_home_base**. This simple piece of code is called a **Getter** and is used to access attributes of the class in a formal way, and in doing so prevents unauthorised access to variables.

The getters, setters and methods of a class are collectively called its **Interface**. Users of the class interact with it through its **class interface**, the remainder of the code is inaccessible and maintains the encapsulation and abstraction of the class.


**When Are Getters & Setters Appropriate?**
>**Encapsulation & Data Protection**: If a variable requires controlled access (e.g., preventing invalid values like a negative age), getters and setters help enforce constraints.   
**Maintaining Consistency**: If changing a variable should trigger additional logic (like updating a database or recalculating values), setters ensure structured updates.   
**Future-Proofing Code**: If internal implementation may change but the interface should remain the same, getters/setters provide flexibility without breaking dependent code - the interface remains stable but the implementation can vary.  

**When Are They Not Necessary?**
>**For simple variables**: If an attribute doesn't require validation or extra logic, direct access (self.var) is usually more Pythonic than using explicit getters/setters.  
**Pythonic Style Encourages Direct Access**: Unlike Java, Python doesn’t enforce private variables, so many Python developers prefer property decorators (@property) over traditional getter/setter methods. These will be discussed later.  
**Overuse Can Make Code Verbose**: Adding getters/setters for variables that don’t need constraints can unnecessarily complicate code.

**`__str__`**

The `__str__()` method automatically executes when an object is converted to a string e.g. when you print.   
Recall that in the early stages of this notebook we asked you to print the three objects you created of "TheSimplestClass" and all that was presented was a memory location of the following form:

><__main__.TheSimplestClass object at 0x00000288C5EC0F10>

This is not particularly useful and it would be better to print something that would be useful to the programmer and that would tell them something about the class. This would generally be a list of variables so that `__str__` can act as a bit of a help function.
It should be a brief summary of the system variables that indicate its current state i.e. the latest values of all the variables and can be useful in debugging.

Run the code below to see how `__str__` improves the value of a printed object:

```python
class Person:
    def __init__(self, forename, surname, age):
        self.forename = forename
        self.surname = surname
        self.age = age

    def __str__(self):
        return f"Name: {self.forename} {self.surname}, Age: {self.age}"

# Creating an instance
person1 = Person("Alice", "Bobtail", 28)

# Printing the object which executes: __str__
print(person1)  
# Output: Alice Bobtail, Age: 28
```

**`__dict__`**

Despite the value of the `__str__` method shown above, python users can also get access to another of the *dunder_methods* (double underscore methods) called **`__dict__`**.   
When executed, it prints a *dictionary* of the class variables - in typical dictionary style.
Its use is shown below:

```python
class Person:
      def __init__(self, forename, surname, age, unit):
        
        self.forename = forename
        self.surname = surname
        self.age = age
        self.unit = unit

        def __str__(self):
            return f"{self.forename} {self.surname}, Age: {self.age}, Unit: {self.unit}"

person1 = Person("Alice", "Bobtail", 28, "Cosford")

# using represent
print(person1.__dict__)   # Output: Represents a person with attributes like name, age, and unit.
```

Given its similarity to our use of `__str__`, you may wonder why we bother to configure `__str__` in the ways we suggest, `__dict__` does this for us?. The reason is that `__str__` can be formatted in much better ways and even handle *missing* variables. Run the code below to see what can be achieved using `__str__` which cannot be done with `__dict__` , note however, the use of `__dict__` within the `__str__` method :

```python
class Car:
    def __init__(self, make, model, year, color, mileage=None):
        self.make = make
        self.model = model
        self.year = year
        self.color = color
        self.mileage = mileage  # Optional attribute

    # You are not expected to produce anything so elaborate in this course.
    def __str__(self):
        attributes = '\n'.join(f"{key.capitalize()}: {value if value is not None else 'Unknown'}" 
                               for key, value in self.__dict__.items())
        return f"\n--- Car Details ---\n{attributes}\n------------------"

# Example usage
my_car = Car("Toyota", "Corolla", 2020, "Blue")
print(my_car)

another_car = Car("Ford", "Fiesta", 2018, "Red", mileage=45000)
print(another_car)
```

**`__doc__` strings**

A doc-string is a description of the overall operationn of a:
>Module  
Class  
function  

that will be useful to developers. They are written between *treble single quotes,* just after the declaration in each case.  
Depending on the type of development environment being used they appear as *pop-ups* after hovering over the various names. Depending on the system, they may be named as **tooltips, hover-tooltips** (pyCharm, VSCode) or **help-text** or **quick-info** in others. Regardles of their name, they will contain the information printed between the triple quotes. They can be accessed like other methods in a class using dot notation.

if you ever need to see the doc string method for a single class, you can access it using:

> print(Person.`__doc__`)

```python
class Person:
    """Represents a person with attributes like name, age, and unit."""

    def __init__(self, forename, surname, age, unit):
        """Initializes a Person with forename, surname, age, and unit."""
        self.forename = forename
        self.surname = surname
        self.age = age
        self.unit = unit

    def __str__(self):
        """Returns a readable string representation of the Person object."""
        return f"{self.forename} {self.surname}, Age: {self.age}, Unit: {self.unit}"

# Retrieving docstrings
print(Person.__doc__)   # Output: Represents a person with attributes like name, age, and unit. - the class doc string
print(Person.__init__.__doc__)   # Output: Initializes a Person with forename, surname, age, and unit - the __init__ doc string
print(Person.__str__.__doc__)    # Output: Returns a readable string representation of the Person object - the __str__ doc string
```
