## A brief introduction

**What is MatPlotLib?**  

Matplotlib is a powerful data visualisation library in Python. It allows you to create:  

>Line plots - graphs (trends over time):  
>Bar charts & histograms (categorical data)  
>Scatter plots (to explore relationships)  
>3D plots (for more advanced visualisations)  


**Installing matplotlib**

Before using Matplotlib, it needs to be installed using pip or using pycharms package loader in the normal way:

> pip install matplotlib


With MatPlotLib installed, it's now possible to create a simple plot using a small set of example data held in two lists:

**Example data**

Data for a plot need not always be provided in a list, it can originate from:

> Lists: Standard Python lists [1, 2, 3, 4]   
> NumPy Arrays: More efficient than lists (np.array[1, 2, 3, 4]) - recall standard python has no provision for arrays    
> Pandas Series & DataFrames: for structured data df['column_name']   
> Tuples: Works similarly to lists (1, 2, 3, 4)  
> Generators & Iterables: Can pass dynamically generated data

Lets examine how these and other features can be used in a more advanced plot:


```python
import matplotlib.pyplot as plt
import numpy as np

# Generate data

# np.linespace is a NumPy function used to create an array of evenly spaced values
# between a starting and ending point. It’s very useful for plotting, numerical
# simulations, and generating sample data.

# numpy.linespace(start, stop, num = 50)
# start = first value in the sequence
# stop = last value in the sequence
# num = the number of evenly space values (default = 50)

# 250 points from 0 to 10
x = np.linspace(0, 10, 250)
y1 = np.sin(x)
y2 = np.cos(x)

# Create the plot - size is in inches
plt.figure(figsize=(8, 5))

# Plot and style the first line (sine wave)
plt.plot(x, y1, linestyle='--', color='tomato', label='Sine Wave')

# and the second line (cosine wave)
plt.plot(x, y2, linestyle='-', color='royalblue', label='Cosine Wave')

# Add labels and a title - with some style changes
plt.xlabel("X Values", fontsize=10)
plt.ylabel("Y Values", fontsize=10)
plt.title("Stylized Sine & Cosine Waves", fontsize=10, fontweight='bold')

# Add the grid and a legend
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.7)
plt.legend()

# Adding annotation
plt.annotate("Peak", xy=(np.pi/2, 1), xytext=(2, 1.2), arrowprops=dict(facecolor='black', arrowstyle='->'))

# Finally - Show the plot
plt.show()
```

This may be absolutely perfect for many uses but in engineering, log scales are used to scale very large ranges of data.

# Logarithmic axes

In this example, logarithmic axes are introduced - these are useful when dealing with data that spans multiple orders of magnitude—such as scientific measurements, engineering data or financial trends.

```python
import matplotlib.pyplot as plt
import numpy as np

# 100 values from 1 to 100
x = np.linspace(1, 100, 100)

# Exponential growth in Y
y = x**2  

plt.plot(x, y, color='royalblue')
plt.yscale('log')  # Set logarithmic scale for Y-axis

plt.xlabel("X Values")
plt.ylabel("Y (log scale)")
plt.title("Logarithmic Y-Axis Example")
plt.grid(True)

plt.show()
```

Notice how the mathematical function y = x^2 is used, the plot(x, y) function calculates a value of y for each of the 100 points in x. It is not supplied as a list as was the first example. Which method would you prefer?

The log scales are invoked using:  

>plt.xscale('log')  
>plt.yscale('log')

# Subplots   

Subplots refer to multiple charts in a single figure.   
You should take note of how these are organised - in the code below, it occurs in the line:

>fig, axes = plt.subplots(2, 1, figsize=(8, 6))

where it:

>Creates a figure (fig): The overall container holding the subplots  
>Creates multiple axes (axes): Individual plots inside the figure  
>Specifies 2, 1: Arranges plots in 2 rows & 1 column  
>Sets figsize=(8, 6): Defines the width (8) and height (6) of the **entire** figure - in inches!  

**Subplot layout**


<img src="images\matplotlib_subplots.png" width="450"/>    


```python
import matplotlib.pyplot as plt
import numpy as np

# Create the subplot space
fig, axes = plt.subplots(2, 1, figsize=(6, 3))

# Generate the data - 100 points from 0 to 10
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Index of first plot is [0] - plot and format
axes[0].plot(x, y1, color='blue')
axes[0].set_title("Sine Wave")

# plot and format the second using index [1]
axes[1].plot(x, y2, color='red')
axes[1].set_title("Cosine Wave")

# Adjust layout and display plots
plt.tight_layout() 
plt.show()
```

Whilst this is no substitute for the matplotlib documentation - lets finish with an example of a 3-D plot:

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a mesh grid
X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a figure and 3D axis
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis')

# Labels and title
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("Fancy 3D Surface Plot")

plt.show()
```
